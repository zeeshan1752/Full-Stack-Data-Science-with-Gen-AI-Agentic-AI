## 1. What is Statistics?

Statistics is the process of collecting, organizing, summarizing, analyzing, interpreting, and communicating information from data. It helps us turn observations into evidence for decisions.

## 2. Population and Sample

A population is the complete group of interest. A sample is a subset selected for analysis.

![Population and Sample](https://media.geeksforgeeks.org/wp-content/cdn-uploads/20220302154028/Group-13.jpg)

Source: [GeeksforGeeks](https://www.geeksforgeeks.org/maths/population-and-sample-statistics/)

````python
population_size = 10000
sample_size = 500

print("Population size:", population_size)
print("Sample size:", sample_size)
print("Sample proportion of population:", sample_size / population_size)
````

## 3. Parameter vs Statistic

A **parameter** describes a population. A **statistic** describes a sample.

For example, if the true average height of all students is μ and the average height of 500 sampled students is x̄, then μ is a parameter and x̄ is a statistic.

````python
import statistics

population = [60, 65, 70, 75, 80, 85]
sample = [65, 75, 85]

print("Population mean (illustrative parameter):", statistics.mean(population))
print("Sample mean (statistic):", statistics.mean(sample))
````

## 4. Data and Variables

A **variable** is a characteristic that can take different values. Examples include age, height, marks, city, and department.

A single recorded value is an observation of a variable.

````python
students = [
    {"name": "A", "age": 20, "marks": 78, "department": "CSE"},
    {"name": "B", "age": 21, "marks": 85, "department": "ECE"},
    {"name": "C", "age": 20, "marks": 91, "department": "CSE"}
]

print("Number of observations:", len(students))
print("Variables:", list(students[0].keys()))
````

## 5. Qualitative vs Quantitative Data

![Types of Data](https://media.geeksforgeeks.org/wp-content/uploads/20260818105623706789/types_of_data.webp)

- **Qualitative/categorical:** categories such as city, color, department.
- **Quantitative:** numerical quantities such as age, height, marks.

Source: [GeeksforGeeks](https://www.geeksforgeeks.org/data-science/statistics-the-foundation-of-data-science/)

````python
qualitative = ["CSE", "ECE", "CSE", "Mechanical"]
quantitative = [78, 85, 91, 74]

print("Qualitative example:", qualitative)
print("Quantitative example:", quantitative)
````

## 6. Discrete vs Continuous Data

**Discrete:** countable values, such as number of students or number of defects.

**Continuous:** measured values that can take many values in a range, such as height, weight, time, and temperature.

````python
discrete = [10, 11, 12, 13]
continuous = [170.1, 170.15, 170.157, 170.1572]

print("Discrete values:", discrete)
print("Continuous measurements:", continuous)
````

## 7. Four Levels of Measurement

![Four Levels of Measurement](https://media.geeksforgeeks.org/wp-content/uploads/20251206123755209876/the_four_level_of_measu.webp)

The four levels are:

1. **Nominal** – categories only
2. **Ordinal** – categories with order
3. **Interval** – equal intervals, no true zero
4. **Ratio** – equal intervals and a true zero

Source: [GeeksforGeeks](https://www.geeksforgeeks.org/data-science/statistics-the-foundation-of-data-science/) and [OpenStax](https://openstax.org/books/introductory-statistics-2e/pages/1-3-frequency-frequency-tables-and-levels-of-measurement)

### Classification examples

| Variable | Level | Why? |
|---|---|---|
| Blood group | Nominal | Labels, no ranking |
| Satisfaction | Ordinal | Ordered categories |
| Celsius temperature | Interval | Equal differences, no true zero |
| Weight | Ratio | Equal differences and meaningful zero |
| Student ID | Nominal | Number is only a label |

## 8. Sampling Methods

Common methods include:

- simple random sampling
- systematic sampling
- stratified sampling
- cluster sampling
- convenience sampling

OpenStax gives detailed explanations of these methods and emphasizes that sample quality affects the reliability of conclusions.

Reference: [OpenStax – Data, Sampling, and Variation](https://openstax.org/books/introductory-statistics-2e/pages/1-2-data-sampling-and-variation-in-data-and-sampling)

````python
population = list(range(1, 21))

# Simple random sample: choose 5 observations
import random
random.seed(42)
sample = random.sample(population, 5)

print("Population:", population)
print("Random sample:", sample)
````

## 9. Systematic Sampling

Choose a starting point and then select every k-th observation.

Example: starting at 3 and taking every 4th item gives 3, 7, 11, 15, ...

````python
population = list(range(1, 21))
start = 3
k = 4
systematic_sample = population[start-1::k]

print("Systematic sample:", systematic_sample)
````

## 10. Stratified Sampling

Divide the population into meaningful subgroups (strata), then sample within each group.

Example: a university may divide students by department and sample students from each department so that every department is represented.

````python
students = [
    ("A", "CSE"), ("B", "CSE"), ("C", "ECE"),
    ("D", "ECE"), ("E", "ME"), ("F", "ME")
]

strata = {}
for student, department in students:
    strata.setdefault(department, []).append(student)

for department, members in strata.items():
    print(department, "->", members)
````

## 11. Cluster Sampling

Divide the population into clusters, randomly choose some clusters, and study the selected clusters.

Example: randomly selecting several schools and surveying students within those selected schools.

## 12. Convenience Sampling

Choose observations because they are easy to access. It is fast, but it can introduce bias.

Example: asking only your friends about the opinion of all university students.

## 13. Frequency

Frequency is the number of times a value occurs.

````python
marks = [60, 70, 70, 80, 80, 80]

frequency = {}
for mark in marks:
    frequency[mark] = frequency.get(mark, 0) + 1

print(frequency)
````

## 14. Frequency Distribution

A frequency distribution groups observations and records how often they occur. Detailed frequency tables, histograms, and related calculations will be covered in **02 Descriptive Statistics**.

## 15. Sampling Error vs Sampling Bias

**Sampling error** is natural variation caused by studying a sample rather than the entire population.

**Sampling bias** is a systematic problem in the selection process that makes the sample unrepresentative.

Remember:

```text
Sampling error → expected random variation
Sampling bias  → systematic selection problem
```

## 16. Basic Outlier Detection

An outlier is an unusually distant observation. Do not delete it automatically; first investigate whether it is an error or a genuine rare observation.

````python
values = [12, 13, 14, 15, 16, 17, 95]
mean_value = sum(values) / len(values)

print("Mean:", round(mean_value, 2))
print("Potentially unusual value:", 95)
````
