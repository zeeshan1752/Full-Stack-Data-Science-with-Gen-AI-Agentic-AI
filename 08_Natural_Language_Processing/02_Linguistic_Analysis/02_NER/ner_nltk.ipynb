# Named Entity Recognition with NLTK

NER detects entity spans and assigns categories such as person, organization, or location. In this NLTK example, tokenization and POS tagging feed the named-entity chunker.

```python
import nltk

sentence = "Ada Lovelace worked in London and wrote about the Analytical Engine."
tokens = nltk.word_tokenize(sentence)
tagged_tokens = nltk.pos_tag(tokens)
entity_tree = nltk.ne_chunk(tagged_tokens, binary=False)
print(entity_tree)
```
