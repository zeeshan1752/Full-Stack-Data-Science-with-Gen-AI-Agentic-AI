# Manual Bag of Words

## Steps

1. Normalize case and tokenize.
2. Build one shared vocabulary.
3. Count each vocabulary term in each document.

The approach ignores word order and semantic similarity.

```python
import re
```

```python
from collections import Counter

documents = [
    "cats chase small mice",
    "dogs chase cats",
    "mice like small spaces",
]
```

```python
tokenized = [re.findall(r"[a-z]+", doc.lower()) for doc in documents]
```

```python
vocabulary = sorted({word for doc in tokenized for word in doc})
```

```python
matrix = [[Counter(doc)[word] for word in vocabulary] for doc in tokenized]
```

```python
print("Vocabulary:", vocabulary)
```

```python
for row in matrix:
    print(row)
```
