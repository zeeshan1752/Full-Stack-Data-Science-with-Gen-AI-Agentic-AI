# Word2Vec with Gensim

```python
import re
from gensim.models import Word2Vec
```

```python
import sys
print("Python path:", sys.executable)

from gensim.models import Word2Vec

print("Gensim Word2Vec is ready!")
```

```python
sentences = [
    "cats chase small mice",
    "dogs chase small cats",
    "mice hide in small spaces",
    "dogs and cats are animals",
]

sentences
```

```python
tokenized = [re.findall(r"[a-z]+", line.lower()) for line in sentences]
model = Word2Vec(
    sentences=tokenized,
    vector_size=30,
    window=2,
    min_count=1,
    workers=1,
    seed=42,
    epochs=100,
)
```

```python
print("Vector shape:", model.wv["cats"].shape)
```

```python
print("Vocabulary:", sorted(model.wv.index_to_key))
```

## CBOW and Skip-gram comparison

```python
comparison_parameters = {
    "vector_size": 30,
    "window": 2,
    "min_count": 1,
    "workers": 1,
    "seed": 42,
    "epochs": 100,
}

cbow_model = Word2Vec(
    sentences=tokenized,
    sg=0,
    **comparison_parameters,
)

skipgram_model = Word2Vec(
    sentences=tokenized,
    sg=1,
    **comparison_parameters,
)

word = "cats"
print("CBOW vector shape:", cbow_model.wv[word].shape)
print("Skip-gram vector shape:", skipgram_model.wv[word].shape)
print("CBOW similar words:", cbow_model.wv.most_similar(word, topn=3))
print("Skip-gram similar words:", skipgram_model.wv.most_similar(word, topn=3))
```
