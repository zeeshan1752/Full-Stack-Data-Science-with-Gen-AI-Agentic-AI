# Stemming

Stemming applies rules to reduce related word forms to a stem. The result may not be a dictionary word. Porter, Lancaster, and Snowball are distinct algorithms; their results can differ.

The original Stemming_01.py and Stemming_02.py remain unchanged. The companion examples below compare the three algorithms. Install NLTK with python -m pip install nltk.

```python
from nltk.stem import LancasterStemmer, PorterStemmer, SnowballStemmer

words = ["connect", "connected", "connecting", "connection"]
stemmers = {
    "Porter": PorterStemmer(),
    "Lancaster": LancasterStemmer(),
    "Snowball": SnowballStemmer("english"),
}

for name, stemmer in stemmers.items():
    print(name)
    for word in words:
        print(f"  {word} -> {stemmer.stem(word)}")
```

## What to notice

Compare the output across algorithms. A stemmer is fast and rule-based, but it can produce a shortened form that is not a dictionary word.
