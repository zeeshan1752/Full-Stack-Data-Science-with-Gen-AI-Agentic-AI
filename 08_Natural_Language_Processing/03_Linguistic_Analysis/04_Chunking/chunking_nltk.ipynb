# Chunking and Chinking with NLTK

Chunking groups tagged tokens into shallow phrases such as noun phrases and prepositional phrases. Chinking removes a selected pattern from an existing chunk. This is syntactic phrase chunking; it is different from splitting a long document into model-sized token windows.

## Prepare tokens and part-of-speech tags

This example follows the trainer's rule-based NLTK lesson. If NLTK reports a missing tokenizer or tagger resource, install the resource named in the error using the NLTK data guide.

```python
import nltk
from nltk import RegexpParser, pos_tag, word_tokenize

text = "The quick brown fox jumps over the lazy dog."
tokens = word_tokenize(text)
tagged_tokens = pos_tag(tokens)

print(tagged_tokens)
```

## Chunk noun, verb, and prepositional phrases

The grammar first marks noun phrases, then prepositional phrases, and finally verb phrases. The rules are hand-written, so they depend on the quality of the part-of-speech tags.

```python
chunk_grammar = r"""
NP: {<DT>?<JJ.*>*<NN.*>+}
PP: {<IN><NP>}
VP: {<VB.*><NP|PP>*}
"""

chunk_parser = RegexpParser(chunk_grammar)
chunked_tree = chunk_parser.parse(tagged_tokens)
print(chunked_tree)
```

## Chink selected tags out of a chunk

This second grammar begins with one broad chunk and removes verbs and prepositions from it. A chink rule is written between closing and opening braces: }pattern{.

```python
chink_grammar = r"""
CHUNK: {<.*>+}
       }<VB.*|IN>{
"""

chink_parser = RegexpParser(chink_grammar)
chinked_tree = chink_parser.parse(tagged_tokens)
print(chinked_tree)
```

## Notes

- NLTK's RegexpParser is a rule-based shallow parser; it does not build a full dependency or constituency parse.
