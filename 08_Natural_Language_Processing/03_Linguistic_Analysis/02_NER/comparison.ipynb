# Named entity recognition comparison: NLTK and spaCy
 Both implementations use `extract_entities(text)` and return a list of `(entity_text, label)` pairs. This notebook uses the same sentence from the shared `data/` folder.
 Install NLTK and spaCy, download `en_core_web_sm`, and install any NLTK resources requested at runtime.


```python
from pathlib import Path
 import sys
 from time import perf_counter
 
 cwd = Path.cwd().resolve()
 nlp_root = next(path for path in [cwd, *cwd.parents] if (path / "data" / "sample_texts.txt").is_file())
 topic_dir = nlp_root / "03_Linguistic_Analysis" / "02_NER"
 sys.path.insert(0, str(topic_dir))
 
 from ner_nltk import extract_entities as extract_entities_nltk
 from ner_spacy import extract_entities as extract_entities_spacy
 
 sample_text = next(
     line for line in (nlp_root / "data" / "sample_texts.txt").read_text(encoding="utf-8").splitlines()
     if line.startswith("Ada Lovelace")
 )
 sample_text
```

```python
results = {}
 for name, extractor in [(
     "NLTK", extract_entities_nltk),
     "spaCy", extract_entities_spacy,
 ]:
     started = perf_counter()
     entities = extractor(sample_text)
     results[name] = {"entities": entities, "elapsed_seconds": perf_counter() - started}
 
 results
```

## Comparison note
 Speed: first-call timing includes spaCy model startup and NLTK resource loading, so use repeated runs for a fairer steady-state comparison. Accuracy: the two models can return different spans or labels; one sentence is not enough to rank them. Ease of use: NLTK combines tokenization, POS tagging, and chunking steps, while spaCy returns entities from one document pipeline.
