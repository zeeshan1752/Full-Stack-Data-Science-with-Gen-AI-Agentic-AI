# Word2Vec with Gensim

Word2Vec learns word vectors from local context. The small corpus below is only a coding demonstration; it is not enough data for useful general-purpose embeddings.

## Dependency

Install Gensim if needed with python -m pip install --upgrade gensim. The original practice notebook records a missing-Gensim import error, so this new companion has not been represented as an executed result.

```python
import re
from gensim.models import Word2Vec
```

```python
import sys
print("Python path:", sys.executable)

from gensim.models import Word2Vec

print("Gensim Word2Vec is ready!")
```

```python
sentences = [
    "cats chase small mice",
    "dogs chase small cats",
    "mice hide in small spaces",
    "dogs and cats are animals",
]

sentences
```

```python
tokenized = [re.findall(r"[a-z]+", line.lower()) for line in sentences]
model = Word2Vec(
    sentences=tokenized,
    vector_size=30,
    window=2,
    min_count=1,
    workers=1,
    seed=42,
    epochs=100,
)
```

```python
print("Vector shape:", model.wv["cats"].shape)
```

```python
print("Vocabulary:", sorted(model.wv.index_to_key))
```
