# Bag of Words with CountVectorizer

Bag of Words represents each document using vocabulary-term counts. This notebook demonstrates scikit-learn's CountVectorizer, which builds a shared vocabulary and a sparse document-term matrix.

## Create count features

Each row represents a document and each column a vocabulary term. The matrix is sparse because most documents contain only a small portion of a vocabulary.

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "cats chase small mice",
    "dogs chase cats",
    "mice like small spaces",
]
```

```python
vectorizer = CountVectorizer()
matrix = vectorizer.fit_transform(documents)
terms = vectorizer.get_feature_names_out()
```

```python
print("Vocabulary:", terms.tolist())
```

```python
print("Matrix shape:", matrix.shape)
```
