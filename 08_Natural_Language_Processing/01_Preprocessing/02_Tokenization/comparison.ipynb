# Tokenization comparison: NLTK and spaCy
 Use both implementations on the same shared sample. The comparison records token lists and a simple elapsed time; the spaCy model load affects its first call.


```python
from pathlib import Path
 import sys
 from time import perf_counter
 
 cwd = Path.cwd().resolve()
 nlp_root = next(path for path in [cwd, *cwd.parents] if (path / "data" / "sample_texts.txt").is_file())
 topic_dir = nlp_root / "01_Preprocessing" / "02_Tokenization"
 sys.path.insert(0, str(topic_dir))
 
 from tokenization_nltk import tokenize as tokenize_nltk
 from tokenization_spacy import tokenize as tokenize_spacy
 
 sample_text = next(
     line for line in (nlp_root / "data" / "sample_texts.txt").read_text(encoding="utf-8").splitlines()
     if line.startswith("Ada Lovelace")
 )
 sample_text
```

```python
results = {}
 for name, tokenizer in [(
     "NLTK", tokenize_nltk),
     "spaCy", tokenize_spacy,
 ]:
     started = perf_counter()
     tokens = tokenizer(sample_text)
     results[name] = {"tokens": tokens, "elapsed_seconds": perf_counter() - started}
 
 results
```

## Comparison note
 Speed: first-call timing includes spaCy model startup, so repeat runs for a fairer steady-state comparison. Accuracy: token boundaries can differ around punctuation and contractions; review examples from your own text. Ease of use: NLTK has a lightweight tokenizer API, while spaCy provides tokenization as part of a broader NLP pipeline.
