# Tokenization

Tokenization converts text into tokens. Sentence tokenization finds sentence boundaries; word tokenization separates words and punctuation. Different tokenizers can produce different results for contractions and punctuation.

```python
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize

text = "NLP is useful. Tokenizers split text into units!"
sentences = sent_tokenize(text)
words = word_tokenize(text)
word_punctuation = wordpunct_tokenize(text)

print("Sentences:", sentences)
print("Words:", words)
print("Word-punctuation:", word_punctuation)
```

## Additional NLTK tokenizers from trainer material

NLTK also includes blank-line and whitespace tokenizers. They are useful for comparing how token boundaries change with the chosen rule.

```python
from nltk.tokenize import WhitespaceTokenizer, blankline_tokenize

paragraph_text = (
    "The first paragraph has punctuation.\n"
    "It also has a second sentence.\n\n"
    "A blank line starts the next paragraph."
)

paragraphs = blankline_tokenize(paragraph_text)
whitespace_tokens = WhitespaceTokenizer().tokenize(paragraph_text)

print("Paragraphs:", paragraphs)
print("Whitespace tokens:", whitespace_tokens)
```
