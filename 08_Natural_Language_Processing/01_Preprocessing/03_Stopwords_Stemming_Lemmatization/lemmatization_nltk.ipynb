# Lemmatization

Lemmatization maps an inflected word to a dictionary base form. WordNetLemmatizer defaults to noun behavior when no part of speech is supplied, so verbs may not reduce as expected.

```python
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words = ["cars", "running", "better"]
```

```python
print("Default noun behavior:")
for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
print("With verb part of speech:")
for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos=wordnet.VERB))
```
