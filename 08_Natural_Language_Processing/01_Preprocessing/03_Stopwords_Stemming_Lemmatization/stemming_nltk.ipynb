# Stemming

Stemming applies rules to reduce related word forms to a stem. The result may not be a dictionary word. Porter, Lancaster, and Snowball are distinct algorithms; their results can differ.

```python
from nltk.stem import LancasterStemmer, PorterStemmer, SnowballStemmer

words = ["connect", "connected", "connecting", "connection"]
stemmers = {
    "Porter": PorterStemmer(),
    "Lancaster": LancasterStemmer(),
    "Snowball": SnowballStemmer("english"),
}

for name, stemmer in stemmers.items():
    print(name)
    for word in words:
        print(f"  {word} -> {stemmer.stem(word)}")
```
