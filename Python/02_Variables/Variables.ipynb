## Rules:

### 1. Variable does not start with numbers (syntax error), but ends with number

```python
7 = a
```

```python
7abc = 5
```

```python
abc3d = 5
abc3d
```

```python
a5 = 10
a5
```

### 2. Variables are case sensitive

```python
a = 5
A = 10
```

```python
a
```

```python
A
```

### 3. Special character is not allowed, except underscore (_)

```python
@var = 10
```

```python
my_name = "zeeshan"
my_name
```
