# Dictionary

A **dictionary** is a built-in Python data structure used to store data in **key-value pairs**.

Dictionaries are:

* Ordered (insertion order is preserved)
* Mutable
* Do not allow duplicate keys
* Allow duplicate values
* Keys must be unique
* Keys must be immutable/hashable
* Values can be of any data type
* Can store different data types
* Can contain nested dictionaries

Syntax:  `variable = {key : value}`

```python
student = {'name' : 'Zeeshan Jamshed', 'age' : 22, 'is_male': True}
student
```

```python
type(student)
```

```python
len(student)
```

- Empty Dictionary

```python
data = {}
data
```

```python
student
```

```python
student.pop('age')
```

- `dictionary.pop(key, default)`

```python
student.pop('name', 'city') # here 'city' is default if 'name' key is not available then print 'city'
```

```python
student.pop('course', 'gender') # 'gender' will not treated as a key 
```

Returns the value associated with a key.

### `copy()`

Creates a **shallow copy** of the dictionary.

```python
student_1
```

```python
student_3 = student_1 # student_3 and student_1 have same dictionary in the memory, no separate dictionary is created
```

```python
student_3
```

```python
id(student_3) == id(student_1)
```

```python
student_2 = student_1.copy() # a new dictionary will be created in the memory
```

```python
student_2
```

```python
id(student_2) == id(student_1)
```

```python
id(student_3) == id(student_2)
```

### `setdefault()`

Returns the value of a key if it exists.

If the key does not exist, it adds the key with the specified default value.

```python
data1 = {
    'name' : 'Zeeshan'
}
```

```python
data1
```

```python
data1.setdefault('name', 'Jamshed') # key already exist, it will not update the existing value and print the existing value
```

```python
data1.setdefault('age', 22) # key does not exist, it will add a new key-value pair in the dictionary
```

```python
data1
```

## Looping Through a Dictionary

### Loop Through Keys

```python
data1
```

```python
for key in data1: # key is a variable, we can use any valid variable
    print(key)
```

```python
for _ in data1:
    print(_)
```

```python
for keys in data1.keys():
    print(keys)
```

### Loop Through Values

```python
data1
```

```python
for values in data1.values(): # values is also a variable, we can use any valid variable here
    print(values)
```

### Loop Through Keys and Values

```python
for keys, values in data1.items(): # keys and values are variables
    print(keys, values)
```
