## 2. `copy()`

- Used to copy one list to another. (Creates a copy of a list)

```python
list2 = list1.copy()
```

```python
list2
```

```python
list1 == list2
```

```python
len(list1) == len(list2)
```

```python
id(list1) == id(list2)
```

```python
id(list1) != id(list2)
```

- It is used to merge or concatenate the list.

```python
a = [1, 2, 3]
b = [4, 5, 6]
```

```python
a.extend(b) # b will me merged in a
```
