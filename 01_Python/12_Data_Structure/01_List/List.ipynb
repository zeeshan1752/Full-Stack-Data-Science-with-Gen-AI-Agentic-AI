# Looping and Built-in Functions with Lists

## 1. Using `for` Loop with a List

A `for` loop can be used to access each element of a list one by one.

````python
unpack
````

````python
for i in unpack: # we can use "value" instead of "i" also because both are variable (we can use any variable here)
    print(i)
````

## 2. Using `for i in range()`

`range()` can be used to loop through the **indexes** of a list.

````python
unpack
````

````python
for i in range(len(unpack)):
    print(i)
````

````python
for i in range(len(unpack)):
    print(unpack[i])
````

## 3. Using `enumerate()`

`enumerate()` is used when you want to get both the **index and value** while looping through a list.

````python
unpack
````

````python
for i in enumerate(unpack):
    print(i)
````

In this case, `i` contains a **tuple**:

```text
(index, value)
```

### Using `enumerate()` with Two Variables

````python
unpack
````

````python
for i, value in enumerate(unpack):
    print(i, value)
````

* `i` → index
* `value` → element at that index

## 4. `all()`

`all()` returns `True` if **all elements** in an iterable are truthy. (It means non-zero items)

````python
all_1 = [1, 2, 5.8, 1+2j, 'Zeeshan', True]
````

````python
all(all_1)
````

````python
all_2 = [1, 2, 5.8, 1+2j, 'Zeeshan', True, ''] # empty string are considered as zero
````

````python
all(all_2)
````

## 5. `any()`

`any()` returns `True` if **at least one element** in an iterable is truthy.

````python
all_1
````

````python
any(all_1) # any non zero element
````

````python
all_2
````

````python
any(all_2)
````

##### If all elements are zeros or empty then it will print false
