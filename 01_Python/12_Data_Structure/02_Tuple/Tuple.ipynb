# Tuple

A **tuple** is a built-in Python data structure used to store multiple values in a single variable.

Tuples are:

* Ordered
* Immutable
* Allow duplicate values
* Can store different data types
* Support indexing and slicing
* Can contain nested data

#### Syntax: `my_tuple = (value1, value2, value3)`

````python
my_tuple = (10, 20, 30, 40)
````

````python
my_tuple
````

## Empty Tuple

````python
t = ()
````

````python
type(t)
````

````python
len(t)
````

## Single element tuple

A comma is required for a single-element tuple.

````python
tup_1 = (10) # without comma it is not a tuple
````

````python
tup_1
````

````python
type(tup_1)
````

````python
tup_1 = (10,)
````

````python
tup_1
````

````python
type(tup_1)
````

## Indexing

````python
fruits = ('banana', 'guava', 'apple', 'orange', 'litchi')
````

````python
fruits[0]
````

````python
fruits[4]
````

````python
fruits[-3]
````

## Slicing

````python
fruits[:]
````

````python
fruits[2:3]
````

````python
fruits[-2:-1]
````

## Step slicing

````python
fruits[::-1] # reverse the tuple
````

````python
fruits[1:4:2]
````

## Tuple Methods

Tuples have only **two methods**: `count()` and `index()`

## `count()`

Returns the number of times a value occurs.

````python
num = (10, 20, 10, 10, 20, 37, 43)
````

````python
num
````

````python
num.count(10)
````

````python
num.count(20)
````

## `index()`

Returns the index of the first occurrence of a value.

````python
num
````

````python
num.index(10)
````

````python
num.index(20)
````

````python
num.index(37)
````

# Immutability

Tuples cannot be changed after creation.

````python
t1 = (10, 20, 30, 40)
````

````python
t1[0] = 100 # immutable
````

````python
t1
````

````python
t1.count(10)
````

Tuples do not support methods such as:

```python
append()
remove()
pop()
clear()
```

````python
t1.append(10)
````

````python
t1.remove(10)
````

````python
t1.pop()
````

````python
t1.clear()
````

## Adding elements

You cannot directly add an element to an existing tuple.

However, you can create a **new tuple** using concatenation:

````python
tup_2 = (1, 2, 3)
````

````python
tup_2 = tup_2 + (4,)
````

````python
tup_2
````

## Deleting a Tuple

The entire tuple can be deleted using `del`.

**Note:** `del` is **not a tuple method**. It is a Python statement.

Individual element cannot be deleted.

````python
del tup_2
````

````python
tup_2
````
