# Set

A **set** is a built-in Python data structure used to store multiple values.

Sets are:

* Unordered
* Mutable
* Do not allow duplicate values
* Can store different data types
* Written using curly braces `{}`

- An empty set must be created using `set()`:

- `{}` creates an empty dictionary, not a set.

```python
s = {}
s
```

```python
type(s) # {} - always considered as dictionary, for empty set s = set()
```

```python
s=set() # empty set
print(type(s))
```

# Set Operation

## `union()`

Combines all unique elements from both sets.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
c = {8, 9, 10}
```

```python
a.union(b)
```

```python
a | b | c # both pipe symbol '|' and .union() can be used
```

### `intersection()`

Returns elements that are common to both sets.

```python
print(a)
print(b)
print(c)
```

```python
a.intersection(b)
```

```python
a.intersection(c)
```

```python
b.intersection(c)
```

```python
a & b # both .intersection() and '&' is used for intersection of sets
```

```python
b & c
```

```python
a & b & c
```

### `difference()`

Returns elements that are present in the first set but not in the second.

```python
print(a)
print(b)
print(c)
```

```python
a.difference(b)
```

```python
b.difference(a)
```

```python
b.difference(c)
```

```python
b - c # both .difference and '- (minus)' can be used for difference of sets
```

### `symmetric_difference()`

Returns elements that are present in either set, but not in both.

```python
print(a)
print(b)
print(c)
```

```python
a.symmetric_difference(b)
```

```python
b.symmetric_difference(a)
```

```python
b ^ c # both .symmetric_difference() and '^' can be used
```

```python
a ^ b
```

### `difference_update()`

Removes elements of another set from the original set.

```python
a1 = {1, 2, 3, 4}
b1 = {3, 4, 5}
```

```python
a1.difference_update(b1)
```

```python
a1 # changed
```

- `difference()` → returns a new set, original set is unchanged.

- `difference_update()` → modifies the original set in place.

```python
b1 # unchanged
```

### `symmetric_difference_update()`

Updates the original set with elements that are present in either set, but not in both.

```python
a2 = {1, 2, 3, 4}
b2 = {3, 4, 5, 6}
```

```python
b2.symmetric_difference_update(a2)
```

```python
b2 # changed
```

- `symmetric_difference()` → returns a new set, original set is unchanged.
  
- `symmetric_difference_update()` → modifies the original set in place.

```python
a2 # unchanged
```

## Set Comparison

```python
a3 = {1, 2, 3, 4, 5, 6, 7, 8, 9}
b3 = {3, 4, 5, 6, 7, 8}
c3 = {10, 20, 30, 40}
```

### `issubset()`

Checks whether all elements of one set are present in another set.

```python
b3.issubset(a3)
```

```python
c3.issubset(b3)
```

- The `<=` operator can also be used:

```python
b3 <= a3
```

```python
c3 <= b3
```

### `issuperset()`

Checks whether a set contains all elements of another set.

```python
print(a3)
print(b3)
print(c3)
```

```python
a3.issuperset(b3)
```

```python
a3.issuperset(c3)
```

- The `>=` operator can also be used:

```python
b3.issuperset(c3)
```

### `isdisjoint()`

Checks whether two sets have no elements in common.

```python
print(a3)
print(b3)
print(c3)
```

```python
a3.isdisjoint(c3)
```

```python
a3.isdisjoint(b3)
```

```python
b3.isdisjoint(c3)
```

## Frozenset

A **frozenset** is an immutable version of a set.

Unlike a normal set, a frozenset cannot be changed after it is created.

```python
frozen = frozenset([1, 2, 3, 4])
```

```python
frozen
```

```python
type(frozen)
```

```python
frozen.add(4)
```

```python
b3
```

```python
a4 = frozenset(b3)
```

```python
a4
```

```python
a4.add(5)
```

Frozensets support operations such as:

* `union()`
* `intersection()`
* `difference()`
* `symmetric_difference()`
* `issubset()`
* `issuperset()`
* `isdisjoint()`

Frozensets do not support methods that modify the set, such as:

* `add()`
* `update()`
* `remove()`
* `discard()`
* `pop()`
* `clear()`

A frozenset can be an element of another set because it is immutable.
