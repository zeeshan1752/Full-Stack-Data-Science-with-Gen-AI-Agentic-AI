# Range

`range()` is a built-in Python function used to generate a sequence of integers.

It is commonly used with `for` loops.

## Syntax

`range(stop)`

`range(start, stop)`

`range(start, stop, step)`

* `start` → starting value
* `stop` → ending limit (**not included**)
* `step` → difference between values

### `range(stop)`

Starts from `0` and uses a step of `1`.

```python
range(5) # stop value never included
```

```python
list(range(0,5))
```

### `range(start, stop)`

```python
range(10,20) # stop value never included
```

```python
list(range(10, 20))
```

### `range(start, stop, step)`

```python
range(10, 50, 5) # stop value never included
```

```python
list(range(10, 50, 5))
```

- Max 3 argument allowed in `range()`

```python
list(range(10, 50, 5, 2)) # max 3 argument it takes
```

## Type of `range()`

`range()` returns a `range` object.

```python
range_1 = range(0,5)
```

```python
type(range_1)
```

```python
range_1
```

- It does not immediately create a list containing all the numbers.

## Converting `range()` to a List

Use `list()` to convert it into a list.

```python
range_1 = range(0,5)
```

```python
range_1
```

```python
list(range_1)
```

- We can also change `range` to other sequence like `tuple`, `set`.

```python
tuple(range_1)
```

```python
set(range_1)
```

```python
dict(range_1)
```

```python
dict(range(1, 5))
```
