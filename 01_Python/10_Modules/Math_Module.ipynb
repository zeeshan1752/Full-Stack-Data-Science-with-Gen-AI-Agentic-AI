# Python Math Module

The `math` module is a built-in Python module that provides mathematical functions and constants.

```python
import math
```

## sqrt()

Used to find the square root of a number.

```python
print(math.sqrt(25))
print(math.sqrt(16))
print(math.sqrt(100))
```

## pow()

Used to calculate the power of a number.

```python
print(math.pow(2, 3))
print(math.pow(5, 2))
print(math.pow(10, 3))
```

## floor()

Returns the largest integer less than or equal to the given number.

```python
print(math.floor(4.8))
print(math.floor(7.9))
print(math.floor(-4.2))
```

## ceil()

Returns the smallest integer greater than or equal to the given number.

```python
print(math.ceil(4.2))
print(math.ceil(7.1))
print(math.ceil(-4.2))
```

## import math as

`as` is used to give an alias to a module.

```python
import math as m

print(m.sqrt(25))
print(m.pow(2, 3))
print(m.floor(4.8))
print(m.ceil(4.2))
```

## from math import

Used to import specific functions from the module.

```python
from math import sqrt, pow, floor, ceil

print(sqrt(25))
print(pow(2, 3))
print(floor(4.8))
print(ceil(4.2))
```

## from math import *

`*` means all available names from the module.

```python
from math import *

print(sqrt(25))
print(pow(2, 3))
print(floor(4.8))
print(ceil(4.2))
```
