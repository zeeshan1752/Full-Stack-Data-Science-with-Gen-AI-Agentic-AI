# 1. Arithmetic operators

Arithmetic operators are used to perform **mathematical operations**.

| Operator | Name           | Example   | Result |
| -------- | -------------- | --------- | ------ |
| `+`      | Addition       | `10 + 5`  | `15`   |
| `-`      | Subtraction    | `10 - 5`  | `5`    |
| `*`      | Multiplication | `10 * 5`  | `50`   |
| `/`      | Division       | `10 / 5`  | `2.0`  |
| `//`     | Floor Division | `10 // 3` | `3`    |
| `%`      | Modulus        | `10 % 3`  | `1`    |
| `**`     | Exponentiation | `2 ** 3`  | `8`    |

### Addition `+`

- The `+` operator is used to add two values.

```python
a = 5
b = 13
c = a+b
print(c)
print(a+b) # Direct add in print function
```

### Subtraction `-`

- The `-` operator is used to subtract one value from another.

```python
a = 5
b = 13
print(a-b)
print(b-a)
```

### Multiplication `*`

- The `*` operator is used to multiply two values.

```python
a = 5
b = 13
c = 6
print(a*b)
print(b*c)
```

### Division `/`

- The `/` operator is used for division.

- The `/` operator returns the division result as a floating-point value.

```python
c = 15
d = 5
print(c/d)
```

### Floor Division `//`

- The `//` operator is used for floor division.

- It returns the quotient after removing the decimal part by flooring the result.

```python
c = 15
d = 5
print(c//d)
```

```python
print(10//3)
```

### Modulus `%`

- The `%` operator returns the **remainder** after division.

```python
c = 15
d = 5
print(c%d) # remainder 0
```

```python
e = 10
f = 4
print(e%f) # remainder 2
```

### Exponential `**`

- The `**` operator is used to calculate the power of a number.

```python
c = 15
d = 5
print(c**d) # c is base and d is power
```

```python
e = 10
f = 4
print(e**f) # e is base and f is power
```

# 2. Assignment Operator

- Assignment operators are used to **assign or update values in variables**.

- The basic assignment operator is: `=`

#### Assignment Operators Table

| Operator | Meaning                 | Same As      |
| -------- | ----------------------- | ------------ |

| `/=`     | Divide and assign       | `a = a / 5`  |
| `//=`    | Floor divide and assign | `a = a // 5` |
| `%=`     | Modulus and assign      | `a = a % 5`  |
| `**=`    | Power and assign        | `a = a ** 5` |


```python
num = 1 # 1 (value) is assigned to the variable num
print(num)
```

```python
num2 = 10
num2 += 2
print(num2)
```

```python
num3 = 15
num3 -= 2
print(num3)
```

```python
num4 = 20
num4 *= 2
print(num4)
```

```python
num5 = 25
num5 /= 3
print(num5)
```

```python
num6 = 30
num6 //= 7
print(num6)
```

```python
num7 = 35
num7 %= 8
print(num7)
```

# 3. Relational Operator

- Relational operators are also called **comparison operators**.

- They are used to **compare two values**.

- The result of a comparison is either: `True` or `False`

| Operator | Meaning                  | Example  |
| -------- | ------------------------ | -------- |
| `==`     | Equal to                 | `5 == 5` |
| `!=`     | Not equal to             | `5 != 3` |
| `>`      | Greater than             | `5 > 3`  |
| `<`      | Less than                | `3 < 5`  |
| `>=`     | Greater than or equal to | `5 >= 5` |
| `<=`     | Less than or equal to    | `3 <= 5` |

```python
m = 100
n = 101
```

```python
m == n # == used for comparison while = used for assignment
```

```python
m != n
```

```python
m > n
```

```python
m < n
```

```python
m >= n
```

```python
m <= n
```

# 4. Logical Operator

Logical operators are used to **combine or modify conditions**.

Python has three basic logical operators:

* `and`
* `or`
* `not`

## `and` Operator

The `and` operator returns `True` only when **both conditions are True**.

### Truth Table

| Condition 1 | Condition 2 | Result |
| ----------- | ----------- | ------ |
| True        | True        | True   |
| True        | False       | False  |
| False       | True        | False  |
| False       | False       | False  |

```python
a = 5
b = 10
```

```python
a < 6 and b > 6 # we can write that in print function also
```

```python
a < 6 and b > 11 
```

```python
a < 4 and b > 6
```

```python
a < 4 and b > 11
```

## `or` Operator

The `or` operator returns `True` when **at least one condition is True**.

### Truth Table

| Condition 1 | Condition 2 | Result |
| ----------- | ----------- | ------ |
| True        | True        | True   |
| True        | False       | True   |
| False       | True        | True   |
| False       | False       | False  |

```python
c = 10
d = 20
```

```python
c > 4 or d > 4
```

```python
c > 4 or d < 4
```

```python
c < 4 or d > 4
```

```python
c < 4 or d < 4
```

## `not` Operator

### Truth Table

| Condition | `not` Condition |
| --------- | --------------- |
| True      | False           |
| False     | True            |

```python
not True
```

```python
not False
```

```python
not (10 > 5)
```

```python
not (10 < 5)
```

```python
c = 10
d = 20
```

```python
not(c > 4 or d > 4)
```

# 5. Unary Operator

A unary operator is an operator that works with **only one operand**.

Unary operators are mainly used when we want to **represent or change the sign of a value**.

The unary `+` represents a positive value.

The unary `-` changes the sign of a value.

```python
num = 10
print(-num)
```

* `-` is the unary operator.
* `num` is the only operand.

```python
num_5 = 10
print(-num_5)
print(+num_5)
```

```python
num_6 = -10
print(-num_6)
print(+num_6)
```

- We usually do not use `+` as a unary operator because it prints the same value or we can say it can not change anything.
