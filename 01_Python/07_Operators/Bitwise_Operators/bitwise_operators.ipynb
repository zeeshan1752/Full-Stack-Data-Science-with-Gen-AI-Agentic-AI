# Bitwise Operators

Bitwise operators are used to perform operations on the **individual bits (0 and 1)** of an integer.

Python provides the following bitwise operators:

| Operator    | Name                     | Symbol |
| ----------- | ------------------------ | ------ |
| AND         | Bitwise AND              | `&`    |
| OR          | Bitwise OR               | `\|`   |
| XOR         | Bitwise XOR              | `^`    |
| Complement  | Bitwise NOT / Complement | `~`    |
| Left Shift  | Left Shift               | `<<`   |
| Right Shift | Right Shift              | `>>`   |

## Bitwise AND `&`

The AND operator compares each bit of two numbers.

````python
12 & 13 #  12 = 1100, 13 = 1101  -- 1100 & 1101 = 1100 -- 12
````

````python
12 & 5
````

## Bitwise OR `|`

The OR operator compares each bit of two numbers.

* If **at least one bit is `1`**, the result is `1`.

* Only `0 | 0` gives `0`.

````python
12 | 13 #  12 = 1100, 13 = 1101  -- 1100 | 1101 = 1101 -- 13
````

````python
12 | 5
````

## Bitwise XOR `^`

XOR means **Exclusive OR**.

* Different bits → `1`

* Same bits → `0`

````python
12^13
````

````python
12^5
````

````python
19^45
````

## Bitwise Complement `~`

The complement operator changes every bit:

```text
0 → 1
1 → 0
```

It is also called **Bitwise NOT**.

### Important Rule

For an integer:

```text
1's complement + 1 = 2's complement
```

And Python's complement operation follows:

```text
~n = -(n + 1)
```

````python
~5 # ~5 = -(5+1)
````

````python
~10
````

# Left Shift `<<`

The left shift operator shifts the bits towards the **left**.

Syntax:

```python
number << positions
```

````python
12<<2 # 12 = 1100, 1100<<2 -- add 2 more zero in right = 110000 = 48
````

````python
12<<1
````

## Right Shift `>>`

````python
12>>2 # remove 2 bit from right side
````

````python
12>>1
````
