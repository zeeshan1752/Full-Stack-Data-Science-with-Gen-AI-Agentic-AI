## Concatenation

- Adding two or more string or string and integer is called concatenation using "+"
- we cannot directly concatenate a string with an integer, we have to typecast the integer to string then we can add.

```python
first_name = "Zeeshan"
last_name = "Jamshed"
print(len(first_name))
print(len(last_name))
```

```python
full_name = first_name + " " + last_name
print(full_name)
```

```python
var1 = 90
var2 = 'Biryani'
print(type(var1))
print(type(var2))
```

```python
print(var1+var2)
```

```python
print(str(var1)+ " " + var2)
```

```python
print("Price of biryani: " + var1)
```

# Slicing

- start = starting index, included.
- stop = ending position, not included.
- The actual last included index is stop - 1 (Last index = stop - 1)

# Forward Slicing

```python
subject = "Python"
subject
```

```python
len(subject)
```

```python
subject[0:3]
```

```python
subject[3:6]
```

- Slicing from beginning to an index

```python
subject[:5]
```

- Slicing till end from any index

```python
subject[2:]
```

- Complete string using slicing

```python
subject[:]
```

#### String slicing with step

- Syntax: string[start:stop:step]
- The step tells Python how many positions to move at a time.

```python
subject[0:5:1] # When we use step = 1 i.e, it will print all the values from start to stop
```

```python
subject[1:6:2] # When we use step = 2 i.e, it will print start and skip start + 1 and then print start + 2 and so on.... i.e, print one and skip next
```

# Backward Slicing

- Backward slicing uses negative indexes and works from the right side.

```python
sub_2 = "Machine Learning"
sub_2
```

```python
len(sub_2)
```

```python
sub_2[-16:-1]
```

```python
sub_2[-10:-9]
```

# Using Both positive and negative index in slicing 

```python
sub_2[-1:1] # But from -1 (last position), you cannot move forward and reach 1
```

```python
sub_2[-4:1] # -4 means 11, from 11 to 1 is not possible
```

```python
sub_2[-12:11]
```

```python
sub_2[1:-1] #1:-1 means from the second character up to, but not including, the last character
```

- Slicing From Beginning

```python
sub_2[:5]
```

- Slicing from any index to end

```python
sub_2[3:]
```

- Complete string using slicing

```python
sub_2[:]
```

# String Slicing with Step

- Syntax: string[start:stop:step]
- The step tells Python how many positions to move at a time.

```python
step = "Slicing"
step
```

```python
len(step)
```

```python
type(step)
```

```python
step[0:7:1] # step = 1, it means, it starts from 0 and 0+step = 1, 2, 3.... (step = 1 means print all the value from start to stop)
```

## Positive or forward step slicing

```python
step[0:7:2] # step = 2, starts from 0, 0+2 = 2, 4, and 6
```

```python
step[0:100:]
```

```python
step[:50:3]
```

## Negative or backward step slicing (Reversing a string)

```python
step[-7:-1:1]
```

```python
step[-7:-1:-1] # -1 step means move backward, but in backward stop should be less than start
```

```python
step[-1:-7:-1]
```

```python
step[-1:-8:-1]
```

- Whenever we use step = -1 it means it moves from backward

### Reverse a string

```python
step[::-1]
```
