# Python input`()` function

The built-in `input()` function reads a line entered by a user and returns it as a string. Executable examples below use fixed sample responses so the whole notebook can run without pausing; the real syntax is shown in the Markdown examples.

````python
def sample_input(prompt, response):
    """Supply repeatable sample input so Run All never waits for a keyboard response."""
    print(f"{prompt}{response}")
    return response
````

````python
name = sample_input('Enter your name:', "Zeeshan")
````

````python
name
````

### Using `bool()` with `input()`

Python does not directly convert user input into a Boolean in the way beginners usually expect.

For example:

`value = bool(input("Enter something: "))`

This does **not** check whether the user typed `True` or `False`.

It checks whether the resulting string is empty or non-empty.

So for Boolean input, we normally use conditions.

Example:

```
answer = input("Do you like Python? ")

if answer == "yes":
    print(True)
else:
    print(False)
```

````python
boolean = bool(sample_input('Enter True or False:', "False"))
````

````python
boolean
````

````python
type(boolean)
````

````python
boolean1 = bool(sample_input('Enter True or False:', "False"))
````

````python
boolean1 # Non empty string always gives true while type conversion, i.e, we have to use condition.
````

## Parsing fixed input safely

Use `ast.literal_eval()` for values written using Python literal syntax, such as numbers, strings, lists, and dictionaries. Unlike `eval()`, it does not execute arbitrary Python expressions. It does not evaluate expressions such as `2 + 3`; for arithmetic, convert the input to a number and use normal operators.

````python
import ast
exp = ast.literal_eval(sample_input("Enter a Python literal: ", "42"))
````

````python
exp
````
