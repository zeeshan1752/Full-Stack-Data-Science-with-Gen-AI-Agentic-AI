# Python input`()` function

The `input()` function is used to **take input from the user** while the Python program is running.

It allows the user to enter some value through the keyboard.

````python
name = input('Enter your name:')
````

````python
name
````

## Syntax

```python

```

The message inside `input()` is called the **prompt**.

The prompt tells the user what they should enter.

````python
age = input('Enter your age')
````

````python
age
````

By default, the `input()` function returns the user's input as a **string (`str`)**.


````python
type(age)
````

````python
type(name)
````

````python
a = input('Enter a number:')
````

````python
print(a)
print(type(a))
````

````python
b = input('Enter first number:')
c = input('Enter second number:')
````

````python
d = b + c # it will not add the numbers because it takes input from the user as a string, we need to use type conversion to add
print(d)
````

This happens because both values are strings.

So Python performs **string concatenation**, not mathematical addition.

### Using `bool()` with `input()`

Python does not directly convert user input into a Boolean in the way beginners usually expect.

For example:

`value = bool(input("Enter something: "))`

This does **not** check whether the user typed `True` or `False`.

It checks whether the resulting string is empty or non-empty.

So for Boolean input, we normally use conditions.

Example:

```
answer = input("Do you like Python? ")

if answer == "yes":
    print(True)
else:
    print(False)
```

````python
boolean = bool(input('Enter True or False:'))
````

````python
boolean
````

````python
type(boolean)
````

````python
boolean1 = bool(input('Enter True or False:'))
````

````python
boolean1 # Non empty string always gives true while type conversion, i.e, we have to use condition.
````

# `map()`

````python
a1, b1, c1 = map(int, input('Enter three numbers:').split()) # Enter all three values at once using space to separate them
````
