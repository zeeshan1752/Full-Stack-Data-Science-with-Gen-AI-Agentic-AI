# Format Method

- The `format()` method can be used to insert values into a string.
  
- Syntax:  `"{}".format(value)`
  
- The `{}` acts as a placeholder for the values.

```python
num_1 = 52
num_2 = 64
num_3 = 75
num_4 = num_1 + num_2 + num_3
```

```python
print("The sum of {} and {} and {} is {}".format(num_1, num_2, num_3, num_4))
```

```python
num_5 = num_4 - num_2 - num_1 - num_3
```

```python
print("The difference of {}, {}, {} and {} is {}".format(num_4, num_2, num_1, num_3, num_5))
```

# F-String

- F-string is another simple way to insert variables inside a string.

- An `f` or `F` is written before the string.

- Syntax:  `f"text {variable}"`

- F-strings are generally easier to read when we have multiple variables.

```python
b = 23
c = 45
d = 87
e = b + c + d
```

```python
print(f"The sum of {b}, {c} and {d} is {e}.")
```

# `end` Parameter

- By default, `print()` moves to a new line after printing.

- The `end` parameter can change what is printed at the end.

- `end=" "` - Instead of moving to a new line, the first `print()` ends with a space.

- The default value of `end` is: `end="\n"` - `\n` means a new line.

```python
print("Hello") # the default value of end is "\n" - move to a new line
print("World")
```

```python
print("Hello", end = " ") # end = "space" adds a space in the end instead of movie to a new line
print("World")
```

```python
print("Hello", end = "") # end = "no space"
print("World")
```

```python
print("Hello", end = " | ") # end = "we can add anything"
print("World")
```

```python
print("Hello", end = "....") # end = "we can add anything"
print("World")
```

```python
print("Hello", end = "\n") # default value of end = "\n" (we can use like this or we can also remove end, it will do the same)
print("World")
```

# `sep` Parameter

- Syntax:  `sep= "anything"`

- The `sep` parameter is used to specify the separator between multiple values in `print()`.

- By default, `print()` uses a space as the separator.

- The default value is: `sep = "space"`

```python
print(10, 20, 30, 40, sep="-")
```

```python
print(10, 20, 30, 40, sep="/")
```

```python
print(10, 20, 30, 40, sep="|")
```

```python
print(10, 20, 30, 40, sep=" | ")
```

```python
print(10, 20, 30, 40, sep=" | ", end="\nThank You")
```
