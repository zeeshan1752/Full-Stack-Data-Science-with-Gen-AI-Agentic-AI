# Python Type Casting

## Convert other data types to integer

### 1. Float to integer 

```python
height = 6.8
print(height)
print(type(height))
```

```python
int(height) #it removes decimal value
```

```python
type(int(height))
```

### 2. Boolean to int

```python
is_male = True # True = 1
type(is_male)
```

```python
int(is_male) # True = 1
```

```python
type(int(is_male))
```

```python
is_female = False
type(is_female)
```

```python
print(int(is_female))
print(type(int(is_female)))
```

### 3. Complex to integer

```python
com_1 = 10 + 20j
type(com_1)
```

```python
int(com_1)
```

### 4. String to integer

```python
str_1 = '5'
str_2 = 'five'
print(type(str_1))
print(type(str_2))
```

```python
print(int(str_1))
print(type(int(str_1)))
```

```python
int(str_2)
```
