# Calculus

This notebook practices limits, derivatives, optimisation, partial derivatives, gradients, integration, and numerical calculus using Python.


## 1. Setup


```python
import numpy as np
import sympy as sp
```

## 2. Functions


```python
def f(x):
    return x**2 + 2

for value in [0, 1, 2, 3]:
    print(value, f(value))
```

## 3. Limits

Evaluate:

$$
\lim_{x\to2}\frac{x^2-4}{x-2}
$$


```python
x = sp.symbols('x')
expr = (x**2 - 4) / (x - 2)
limit_value = sp.limit(expr, x, 2)
print("Limit:", limit_value)
```

## 4. Symbolic Differentiation


```python
f_expr = x**3 + 2*x**2 - 5*x + 4
f_prime = sp.diff(f_expr, x)
f_second = sp.diff(f_expr, x, 2)

print("f(x) =", f_expr)
print("f'(x) =", f_prime)
print("f''(x) =", f_second)
```

## 5. Derivative Using the Product Rule


```python
product_expr = x**2 * sp.sin(x)
print(sp.diff(product_expr, x))
```

## 6. Derivative Using the Chain Rule


```python
chain_expr = (3*x + 1)**4
print(sp.diff(chain_expr, x))
```

## 7. Tangent Line

For:

$$
f(x)=x^2
$$

find the tangent line at $x=2$.


```python
f = x**2
point = 2
slope = sp.diff(f, x).subs(x, point)
y_value = f.subs(x, point)
tangent = y_value + slope * (x - point)

print("Slope:", slope)
print("Function value:", y_value)
print("Tangent line:", sp.expand(tangent))
```

## 8. Critical Points and Optimisation


```python
g = x**2 - 6*x + 5
g_prime = sp.diff(g, x)
g_second = sp.diff(g, x, 2)
critical_points = sp.solve(sp.Eq(g_prime, 0), x)

print("g'(x):", g_prime)
print("Critical points:", critical_points)
for point in critical_points:
    print("x =", point, "value =", g.subs(x, point), "second derivative =", g_second.subs(x, point))
```

## 9. Increasing and Decreasing Intervals


```python
h = x**2
h_prime = sp.diff(h, x)
print("h'(x):", h_prime)
```

## 10. Partial Derivatives


```python
x, y = sp.symbols('x y')
f_xy = x**2 + 3*x*y + y**2

fx = sp.diff(f_xy, x)
fy = sp.diff(f_xy, y)

print("df/dx:", fx)
print("df/dy:", fy)
```

## 11. Gradient


```python
gradient = sp.Matrix([fx, fy])
print("Gradient:")
sp.pprint(gradient)
```

## 12. Hessian


```python
hessian = sp.hessian(f_xy, (x, y))
print("Hessian:")
sp.pprint(hessian)
```

## 13. Directional Derivative

For:

$$
f(x,y)=x^2+y^2
$$

at $(1,2)$ in direction:

$$
\mathbf{u}=\begin{bmatrix}3/5\\4/5\end{bmatrix}
$$


```python
f2 = x**2 + y**2
grad = sp.Matrix([sp.diff(f2, x), sp.diff(f2, y)])
point = {x: 1, y: 2}
u = sp.Matrix([sp.Rational(3, 5), sp.Rational(4, 5)])

value = grad.subs(point).dot(u)
print("Gradient at point:", grad.subs(point))
print("Directional derivative:", value)
```

## 14. Indefinite Integration


```python
integrand = x**2
antiderivative = sp.integrate(integrand, x)
print("Integral:", antiderivative)
```

## 15. Integration by Substitution Example


```python
integrand = 2*x*(x**2 + 1)**3
result = sp.integrate(integrand, x)
print("Result:", result)
```

## 16. Definite Integral


```python
area = sp.integrate(x**2, (x, 0, 2))
print("Definite integral:", area)
```

## 17. Average Value


```python
a, b = 0, 2
integral_value = sp.integrate(x**2, (x, a, b))
average_value = integral_value / (b - a)
print("Average value:", average_value)
```

## 18. Numerical Derivative


```python
def square(x):
    return x**2

x0 = 3.0
h = 1e-5
numerical_derivative = (square(x0 + h) - square(x0 - h)) / (2*h)
exact_derivative = 2*x0

print("Numerical derivative:", numerical_derivative)
print("Exact derivative:", exact_derivative)
```

## 19. Numerical Integration: Trapezoidal Rule


```python
x_values = np.linspace(0, 2, 1000)
y_values = x_values**2

numerical_area = np.trapezoid(y_values, x_values)
exact_area = 8/3

print("Numerical area:", numerical_area)
print("Exact area:", exact_area)
print("Absolute error:", abs(numerical_area - exact_area))
```

## 21. Final Practical Checklist

- Check the function's domain before evaluating limits or derivatives.
- Distinguish a limit from the actual function value.
- Use the chain rule for nested functions.
- Use the product rule for products of functions.
- Use the quotient rule for quotients when appropriate.
- Remember that an indefinite integral requires an arbitrary constant.
- Distinguish signed area from total geometric area.
- Verify symbolic results numerically when practical.
- Use a sufficiently small step for numerical differentiation, but remember that extremely small steps can introduce floating-point effects.
