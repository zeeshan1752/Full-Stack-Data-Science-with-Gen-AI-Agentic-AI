# Probability Distributions — Practical Notebook

This notebook provides practical calculations and simulations for the distributions explained in `README.md`.

## 1. Discrete Probability Distribution

```python
x = [0, 1, 2]
p = [0.2, 0.5, 0.3]

print("Total probability:", sum(p))
print("Expected value:", sum(a*b for a, b in zip(x, p)))
```

## 2. Bernoulli Distribution

```python
p = 0.2

print("P(X=1):", p)
print("P(X=0):", 1-p)
print("Mean:", p)
print("Variance:", p*(1-p))
```

## 3. Binomial Distribution

```python
import math

n = 4
x = 2
p = 0.5

probability = math.comb(n, x) * p**x * (1-p)**(n-x)
print("P(X=2):", probability)
print("Mean:", n*p)
print("Variance:", n*p*(1-p))
```

## 4. Poisson Distribution

```python
import math

lam = 3
x = 2

probability = math.exp(-lam) * lam**x / math.factorial(x)
print("P(X=2):", probability)
```

## 5. Geometric Distribution

```python
p = 0.2
x = 3

probability = (1-p)**(x-1) * p
print("P(X=3):", probability)
```

## 6. Hypergeometric Distribution

```python
N = 10
K = 4
n = 3
x = 2

probability = (math.comb(K, x) * math.comb(N-K, n-x)) / math.comb(N, n)
print("P(X=2):", probability)
```

## 7. Uniform Distribution

```python
a = 0
b = 10

mean = (a+b)/2
variance = (b-a)**2/12
interval_probability = (6-2)/(b-a)

print("Mean:", mean)
print("Variance:", variance)
print("P(2 <= X <= 6):", interval_probability)
```

## 8. Normal Distribution and Z-Score

```python
x = 70
mu = 60
sigma = 5

z = (x-mu)/sigma
print("Z-score:", z)
```

## 9. Exponential Distribution

```python
lam = 0.2
x = 3

cdf = 1 - math.exp(-lam*x)
mean = 1/lam
variance = 1/(lam**2)

print("P(X <= 3):", cdf)
print("Mean:", mean)
print("Variance:", variance)
```

## 10. Simulating Distributions

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
normal_sample = rng.normal(loc=50, scale=10, size=10000)

plt.hist(normal_sample, bins=40)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Simulated Normal Distribution")
plt.show()
```

## 11. Central Limit Theorem Simulation

```python
rng = np.random.default_rng(42)
population = rng.exponential(scale=2, size=100000)

sample_means = []
for _ in range(5000):
    sample = rng.choice(population, size=30, replace=False)
    sample_means.append(sample.mean())

plt.hist(sample_means, bins=40)
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of the Mean")
plt.show()
```
