## 1. Basic Probability

```python
favourable = 3
total = 6

probability = favourable / total
print("Probability:", probability)
```

## 2. Complement Rule

$$P(A^c)=1-P(A)$$

```python
p_a = 0.7
p_not_a = 1 - p_a
print("P(A^c):", p_not_a)
```

## 3. Addition Rule

```python
p_a = 0.5
p_b = 0.4
p_ab = 0.2

p_a_or_b = p_a + p_b - p_ab
print("P(A union B):", p_a_or_b)
```

## 4. Conditional Probability

$$P(A|B)=\frac{P(A\cap B)}{P(B)}$$

```python
p_ab = 0.2
p_b = 0.5

p_a_given_b = p_ab / p_b
print("P(A|B):", p_a_given_b)
```

## 5. Independence

```python
p_a = 0.5
p_b = 0.5

p_ab = p_a * p_b
print("P(A and B) for independent events:", p_ab)
```

## 6. Law of Total Probability

```python
p_a = 0.6
p_b = 0.4
p_d_given_a = 0.02
p_d_given_b = 0.05

p_d = p_d_given_a * p_a + p_d_given_b * p_b
print("P(D):", p_d)
```

## 7. Bayes' Theorem

```python
p_d = 0.01
p_not_d = 0.99
p_positive_given_d = 0.95
p_positive_given_not_d = 0.05

p_positive = (p_positive_given_d * p_d) + (p_positive_given_not_d * p_not_d)
p_d_given_positive = (p_positive_given_d * p_d) / p_positive

print("P(Positive):", p_positive)
print("P(Disease | Positive):", p_d_given_positive)
```

## 8. Random Variable and Expected Value

```python
values = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

expected_value = sum(x * p for x, p in zip(values, probabilities))
print("Expected value:", expected_value)
```

## 9. Variance of a Random Variable

```python
mean = expected_value
variance = sum(((x - mean) ** 2) * p for x, p in zip(values, probabilities))
standard_deviation = variance ** 0.5

print("Variance:", variance)
print("Standard deviation:", standard_deviation)
```

## 10. Permutations and Combinations

```python
import math

n = 5
r = 3

permutations = math.factorial(n) // math.factorial(n-r)
combinations = math.comb(n, r)

print("Permutations:", permutations)
print("Combinations:", combinations)
```

## 11. Binomial Probability

```python
n = 4
x = 2
p = 0.5

probability = math.comb(n, x) * (p ** x) * ((1-p) ** (n-x))
print("P(X=2):", probability)
```

## 12. Coin-Toss Simulation

A simulation can demonstrate how observed proportions behave as the number of trials increases.

```python
import random

trials = 10000
heads = sum(random.choice([0, 1]) for _ in range(trials))
observed_probability = heads / trials

print("Trials:", trials)
print("Observed probability of Heads:", observed_probability)
```
