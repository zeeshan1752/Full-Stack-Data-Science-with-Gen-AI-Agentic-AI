## Pandas Slicing and Indexing

Slicing is used to select particular rows or columns from a DataFrame. The following operations demonstrate different ways of selecting rows and individual columns.

Special method in OOPS concept which is python constructor in it.

# Slicing in Pandas

### Selecting All Rows

`df[:]` uses slicing to select the DataFrame rows. With no start or stop value specified, the complete row range is selected.

```python
df[:]
```

### Selecting a Row Range

`df[10:21]` selects rows using a start and stop position. The stop position is not included in Python slicing.

```python
df[10:21]
```

### Selecting Rows with a Step

This slice uses start, stop, and step values to select rows at regular intervals. It demonstrates how the step value can skip rows.

```python
df[10:150:20]
```

### Selecting Every 10th Row

`df[::10]` selects rows from the beginning using a step of 10. This is useful for taking a regularly spaced sample of rows.

```python
df[::10]
```

### Reversing the Rows

A step of `-1` reverses the row order. This demonstrates reverse slicing in a DataFrame.

```python
df[::-1]
```

### Reverse Slicing with a Step

A negative step can also skip rows while moving backward. Here, the DataFrame is traversed in reverse using a larger step.

```python
df[::-50]
```

## Filter in Pandas

### Reviewing the Data Before Filtering

The first few rows are displayed before applying conditions. Filtering will be used to select rows based on column values.

```python
df.head()
```

### Creating a Filter Condition

This condition checks which rows have an `InternetUsers` value below 3. The result is a Boolean Series that can be used to filter the DataFrame.

```python
df['InternetUsers'] < 3
```

### Filtering Rows by Internet Usage

The Boolean condition is applied to the DataFrame to return only rows where `InternetUsers` is below 3. This demonstrates conditional row selection.

```python
df[df['InternetUsers'] < 3]
```

### Counting Filtered Rows

`len()` is used on the filtered DataFrame to count how many rows satisfy the condition. This gives the size of the selected subset.

```python
len(df[df['InternetUsers'] < 3])
```

### Filtering by Birth Rate

This condition selects rows where `BirthRate` is greater than 40. It demonstrates filtering using a different numerical column.

```python
df[df['BirthRate'] > 40]
```

### Counting Rows with High Birth Rate

The length of the filtered DataFrame gives the number of rows where `BirthRate` is greater than 40.

```python
len(df[df['BirthRate'] > 40])
```

### Applying Multiple Conditions

Two conditions are combined using `&` so that both conditions must be true. This demonstrates filtering with multiple numerical criteria.

```python
df[(df.BirthRate > 40) & (df.InternetUsers < 2)]
```
