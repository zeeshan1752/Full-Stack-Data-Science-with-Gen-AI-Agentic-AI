## 1. Working with Observations

## 2. Population and Sample

Create a population and select a simple sample from it.

```python
import random

population = list(range(1, 101))
sample = random.sample(population, 10)

print("Population size:", len(population))
print("Sample:", sample)
print("Sample size:", len(sample))
```

## 3. Mean

Calculate the arithmetic mean using the formula:

$$\bar{x} = \frac{\sum_{i=1}^{n}x_i}{n}$$

```python
values = [10, 20, 30, 40, 50]

total = sum(values)
n = len(values)
mean = total / n

print("Total:", total)
print("n:", n)
print("Mean:", mean)
```

## 4. Weighted Mean

Use the weighted mean formula:

$$\bar{x}_w = \frac{\sum_{i=1}^{n}w_i x_i}{\sum_{i=1}^{n}w_i}$$

```python
scores = [80, 70, 90]
weights = [2, 3, 5]

weighted_sum = sum(score * weight for score, weight in zip(scores, weights))
weight_sum = sum(weights)
weighted_mean = weighted_sum / weight_sum

print("Weighted mean:", weighted_mean)
```

## 5. Range

```python
values = [12, 15, 20, 27, 30]

data_range = max(values) - min(values)
print("Range:", data_range)
```

## 6. Frequency Distribution

```python
from collections import Counter

scores = [10, 20, 20, 30, 20, 40, 30, 20]
frequency = Counter(scores)

for score, count in sorted(frequency.items()):
    print(score, ":", count)
```

## 7. Relative Frequency and Percentage

$$R = \frac{f}{n}$$

$$P = \frac{f}{n}\times100$$

```python
f = 20
n = 100

relative_frequency = f / n
percentage = relative_frequency * 100

print("Relative frequency:", relative_frequency)
print("Percentage:", percentage, "%")
```

## 8. Stratified Sampling Practice

```python
total_sample = 1000

cse = total_sample * 0.60
ece = total_sample * 0.25
mechanical = total_sample * 0.15

print("CSE:", int(cse))
print("ECE:", int(ece))
print("Mechanical:", int(mechanical))
print("Total:", int(cse + ece + mechanical))
```

## 9. Statistical Notation Practice

For observations $x_1, x_2, \ldots, x_n$, verify the summation and mean manually using Python.

```python
x = [10, 20, 30]

summation = sum(x)
mean = summation / len(x)

print("Σx:", summation)
print("x̄:", mean)
```
