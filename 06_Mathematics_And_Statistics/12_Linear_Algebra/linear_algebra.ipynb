# Linear Algebra

This notebook provides practical calculations for vectors, matrices, matrix multiplication, systems of equations, determinants, inverses, rank, eigenvalues, and SVD.


## 1. Learning Goals

- Create vectors and matrices with NumPy.
- Perform vector arithmetic and dot products.
- Calculate norms and distances.
- Perform matrix operations.
- Solve linear systems.
- Calculate determinants and inverses.
- Find rank, eigenvalues, eigenvectors, and SVD.


```python
import numpy as np
```

## 2. Vectors


```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

print("a:", a)
print("b:", b)
print("a + b:", a + b)
print("a - b:", a - b)
print("3a:", 3 * a)
```

## 3. Dot Product


```python
dot_product = np.dot(a, b)
print("Dot product:", dot_product)
```

## 4. Norm and Unit Vector


```python
x = np.array([3, 4], dtype=float)

norm = np.linalg.norm(x)
unit_vector = x / norm

print("Norm:", norm)
print("Unit vector:", unit_vector)
print("Unit-vector norm:", np.linalg.norm(unit_vector))
```

## 5. Euclidean Distance


```python
p = np.array([3, 4])
q = np.array([1, 1])

distance = np.linalg.norm(p - q)
print("Distance:", distance)
```

## 6. Matrices and Shapes


```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A:\n", A)
print("Shape of A:", A.shape)
```

## 7. Matrix Addition and Subtraction


```python
print("A + B:\n", A + B)
print("A - B:\n", A - B)
```

## 8. Scalar Multiplication


```python
print("3A:\n", 3 * A)
```

## 9. Matrix Multiplication


```python
print("A @ B:\n", A @ B)
print("Element-wise A * B:\n", A * B)
```

## 10. Transpose


```python
print("A transpose:\n", A.T)
```

## 11. Determinant


```python
det_A = np.linalg.det(A)
print("det(A):", det_A)
```

## 12. Matrix Inverse


```python
A_inv = np.linalg.inv(A)

print("A inverse:\n", A_inv)
print("A @ A_inverse:\n", A @ A_inv)
```

## 13. Solve a Linear System

Solve:

$$
2x+y=5
$$

$$
x-y=1
$$


```python
A_system = np.array([
    [2, 1],
    [1, -1]
], dtype=float)

b_system = np.array([5, 1], dtype=float)

solution = np.linalg.solve(A_system, b_system)
print("Solution:", solution)
```

## 14. Rank


```python
rank_matrix = np.array([
    [1, 2],
    [2, 4]
], dtype=float)

print("Rank:", np.linalg.matrix_rank(rank_matrix))
```

## 15. Null Space Exercise

For a small matrix, verify a candidate null-space vector by checking:

$$
A\mathbf{x}=\mathbf{0}
$$


```python
N = np.array([
    [1, 2],
    [2, 4]
], dtype=float)

v = np.array([-2, 1], dtype=float)
print("N @ v:", N @ v)
```

## 16. Orthogonality


```python
u = np.array([1, 2])
v = np.array([2, -1])

print("Dot product:", np.dot(u, v))
```

## 17. Projection


```python
x = np.array([3, 4], dtype=float)
u = np.array([1, 0], dtype=float)

projection = (np.dot(x, u) / np.dot(u, u)) * u
print("Projection:", projection)
print("Orthogonal remainder:", x - projection)
print("Remainder dot u:", np.dot(x - projection, u))
```

## 18. Eigenvalues and Eigenvectors


```python
E = np.array([
    [2, 0],
    [0, 3]
], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(E)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## 19. Verify an Eigenpair


```python
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    vec = eigenvectors[:, i]
    print("Eigenvalue:", lam)
    print("A v:", E @ vec)
    print("lambda v:", lam * vec)
    print()
```

## 20. Singular Value Decomposition


```python
M = np.array([
    [1, 2],
    [3, 4]
], dtype=float)

U, S, VT = np.linalg.svd(M)

Sigma = np.zeros_like(M)
np.fill_diagonal(Sigma, S)

reconstructed = U @ Sigma @ VT

print("U:\n", U)
print("Singular values:", S)
print("V^T:\n", VT)
print("Reconstructed matrix:\n", reconstructed)
```

## 21. Frobenius Norm


```python
frobenius_norm = np.linalg.norm(M, ord="fro")
print("Frobenius norm:", frobenius_norm)
```
