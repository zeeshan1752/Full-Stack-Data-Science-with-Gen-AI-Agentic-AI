# Descriptive Statistics — Practical Notebook

This notebook provides practical calculations for the concepts explained in `README.md`.

## 1. Create a Dataset

```python
data = [4, 5, 6, 6, 7, 8, 10, 12, 15, 20]

print(data)
print("Number of observations:", len(data))
```

## 2. Mean, Median, and Mode

```python
from statistics import mean, median, mode

print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", mode(data))
```

## 3. Range

```python
data_range = max(data) - min(data)
print("Range:", data_range)
```

## 4. Population and Sample Variance

```python
from statistics import pvariance, variance, pstdev, stdev

print("Population variance:", pvariance(data))
print("Sample variance:", variance(data))
print("Population standard deviation:", pstdev(data))
print("Sample standard deviation:", stdev(data))
```

## 5. Quartiles and IQR

NumPy provides percentile functions for calculating position measures.

```python
import numpy as np

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print("Q1:", q1)
print("Median:", q2)
print("Q3:", q3)
print("IQR:", iqr)
```

## 6. Percentiles

```python
for p in [10, 25, 50, 75, 90]:
    print(f"{p}th percentile:", np.percentile(data, p))
```

## 7. Coefficient of Variation

```python
sample_mean = mean(data)
sample_sd = stdev(data)
cv = (sample_sd / sample_mean) * 100

print("Mean:", sample_mean)
print("Sample standard deviation:", sample_sd)
print("Coefficient of variation:", cv, "%")
```

## 8. Frequency Distribution

```python
from collections import Counter

frequency = Counter(data)
for value, count in sorted(frequency.items()):
    print(value, ":", count)
```

## 9. Histogram

```python
import matplotlib.pyplot as plt

plt.hist(data, bins=5)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

## 10. Box Plot

```python
plt.boxplot(data)
plt.ylabel("Value")
plt.title("Box Plot")
plt.show()
```

## 11. Compare Two Datasets

```python
a = [48, 49, 50, 51, 52]
b = [10, 30, 50, 70, 90]

print("Dataset A mean:", mean(a))
print("Dataset B mean:", mean(b))
print("Dataset A standard deviation:", stdev(a))
print("Dataset B standard deviation:", stdev(b))
```
