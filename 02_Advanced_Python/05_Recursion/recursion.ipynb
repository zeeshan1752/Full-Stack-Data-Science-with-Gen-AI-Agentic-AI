# Recursion

Recursion is a programming technique in which a function or method calls itself.

A recursive solution normally has a **base case** and a **recursive case**.

## 1. Basic Recursion

```python
def count_down(n):
    if n == 0:
        return

    print(n)
    count_down(n - 1)

count_down(5)
```

## 2. Print Numbers from 1 to N

```python
def print_numbers(n):
    if n == 0:
        return

    print_numbers(n - 1)
    print(n)

print_numbers(5)
```

## 3. Factorial

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative integers")
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

## 4. Fibonacci

```python
def fibonacci(n):
    if n == 0:
        return 0

    if n == 1:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(10):
    print(fibonacci(i), end=" ")
```

## 5. Fibonacci with Memoization

```python
def fibonacci_memo(n, memo=None):
    if memo is None:
        memo = {}

    if n in memo:
        return memo[n]

    if n == 0:
        return 0

    if n == 1:
        return 1

    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    return memo[n]

print(fibonacci_memo(10))
```

## 6. Sum of Natural Numbers

```python
def sum_n(n):
    if n == 0:
        return 0

    return n + sum_n(n - 1)

print(sum_n(5))
```

## 7. Sum of Digits

```python
def sum_digits(n):
    if n == 0:
        return 0

    return n % 10 + sum_digits(n // 10)

print(sum_digits(1234))
```

## 8. Reverse a String

```python
def reverse_string(text):
    if len(text) <= 1:
        return text

    return reverse_string(text[1:]) + text[0]

print(reverse_string("Python"))
```

## 9. Palindrome

```python
def is_palindrome(text):
    if len(text) <= 1:
        return True

    if text[0] != text[-1]:
        return False

    return is_palindrome(text[1:-1])

print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

## 10. Power of a Number

```python
def power(base, exponent):
    if exponent == 0:
        return 1

    return base * power(base, exponent - 1)

print(power(2, 5))
```

## 11. GCD

```python
def gcd(a, b):
    if b == 0:
        return a

    return gcd(b, a % b)

print(gcd(48, 18))
```

## 12. Count Digits

```python
def count_digits(n):
    n = abs(n)
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)

print(count_digits(12345))
print(count_digits(0))
print(count_digits(-72))
```

## 13. Find Maximum in a List

```python
def find_max(numbers, index=0):
    if not numbers:
        raise ValueError("The list must not be empty")
    if index == len(numbers) - 1:
        return numbers[index]

    current = numbers[index]
    maximum = find_max(numbers, index + 1)

    return max(current, maximum)

numbers = [10, 50, 20, 80, 30]
print(find_max(numbers))
```

## 14. Direct Recursion

```python
def direct_recursion(n):
    if n == 0:
        return

    print(n)
    direct_recursion(n - 1)

direct_recursion(3)
```

## 15. Indirect Recursion

```python
def function_a(n):
    if n <= 0:
        return

    print("A:", n)
    function_b(n - 1)


def function_b(n):
    if n <= 0:
        return

    print("B:", n)
    function_a(n - 1)


function_a(3)
```

## 16. Recursion Limit

Python has a recursion-depth limit. The exact configured value is **environment-dependent** and should not be assumed to be the same on every Windows, Linux, macOS, WSL, Docker, Jupyter, or Conda environment.

Check the value directly with `sys.getrecursionlimit()`.

```python
import sys

limit = sys.getrecursionlimit()

print("Current recursion limit:", limit)
```

### Change the Recursion Limit

```python
import sys

old_limit = sys.getrecursionlimit()

sys.setrecursionlimit(2000)

print("Before:", old_limit)
print("After:", sys.getrecursionlimit())

# Restore the original value.
sys.setrecursionlimit(old_limit)
print("Restored:", sys.getrecursionlimit())
```

### Demonstrate RecursionError

```python
import sys

print("Configured recursion limit:", sys.getrecursionlimit())

depth = 0

def test_depth():
    global depth
    depth += 1
    test_depth()

try:
    test_depth()
except RecursionError:
    print("RecursionError occurred near depth:", depth)
```

The observed depth is not guaranteed to be exactly equal to the configured limit. It can differ because of Python's implementation and the execution environment.

Do not set an extremely high recursion limit just to avoid `RecursionError`. Excessive recursion can consume significant stack/memory resources and may crash the process.

## 17. Recursive Binary Search

Binary search requires a sorted list. It returns the index of the target, or `-1` when the target is not found.

```python
def binary_search(numbers, target, low, high):
    if low > high:
        return -1

    mid = (low + high) // 2

    if numbers[mid] == target:
        return mid
    if target < numbers[mid]:
        return binary_search(numbers, target, low, mid - 1)
    return binary_search(numbers, target, mid + 1, high)

numbers = [10, 20, 30, 40, 50, 60, 70]
print(binary_search(numbers, 50, 0, len(numbers) - 1))
print(binary_search(numbers, 25, 0, len(numbers) - 1))
```

## 18. Recursion with Nested Lists

A recursive function can visit each item in a nested list. When an item is itself a list, the function calls itself for that list.

```python
def print_nested(data):
    for item in data:
        if isinstance(item, list):
            print_nested(item)
        else:
            print(item)

print_nested([1, [2, 3], [4, [5, 6]]])
```

## 19. Recursion vs Iteration

A loop can solve many of the same repetition problems without making recursive function calls.

```python
def count_down_loop(n):
    for number in range(n, 0, -1):
        print(number)

count_down_loop(5)
```

## 20. Edge Cases to Remember

- `count_digits(0)` should return `1`, because zero has one digit.
- Use `abs(n)` when counting or summing digits of a negative integer.
- Recursive binary search requires sorted input.
- Recursive examples need valid input ranges; for example, factorial is defined here for non-negative integers and power expects a non-negative exponent.

```python
def count_digits(n):
    n = abs(n)
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)

print(count_digits(12345))
print(count_digits(0))
print(count_digits(-72))
```

## Important Rules

1. Identify the base case.
2. Make sure the base case is reachable.
3. Make the problem smaller in every recursive call.
4. Return the recursive result when required.
5. Understand the call stack.
6. Consider time complexity.
7. Consider space complexity.
8. Check the recursion limit for deep recursion.
9. Do not increase the limit to hide an infinite-recursion bug.
10. Use iteration when recursion does not provide a clear advantage.
