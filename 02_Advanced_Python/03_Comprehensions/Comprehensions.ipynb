# Comprehensions

List, Dictionary and Set Comprehension examples.


## 1. List Comprehension


```python
numbers = [1, 2, 3, 4, 5]
squares = [x ** 2 for x in numbers]
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [x for x in numbers if x % 2 == 0]
print(even_numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(result)
```

**Remember the difference:**

- `if` after the `for` filters items, so items that fail the condition are left out.
- `if-else` before the `for` expression chooses a value for every item; it does not filter items out.


```python
words = ["python", "java", "html"]
uppercase = [word.upper() for word in words]
print(uppercase)
```

```python
numbers = [1, 2, 3]
letters = ["A", "B"]
result = [(number, letter) for number in numbers for letter in letters]
print(result)
```

```python
matrix = [[1, 2], [3, 4], [5, 6]]
result = [number for row in matrix for number in row]
print(result)
```

## 2. Dictionary Comprehension


```python
numbers = [1, 2, 3, 4, 5]
squares = {x: x ** 2 for x in numbers}
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = {x: x ** 2 for x in numbers if x % 2 == 0}
print(even_numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
result = {x: "Even" if x % 2 == 0 else "Odd" for x in numbers}
print(result)
```

```python
keys = ["name", "age", "city"]
values = ["Zeeshan", 22, "Lucknow"]
person = {key: value for key, value in zip(keys, values)}
print(person)
```

```python
prices = {"apple": 100, "banana": 50, "orange": 80}
new_prices = {key: value * 2 for key, value in prices.items()}
print(new_prices)
```

```python
marks = {"Python": 85, "Java": 65, "SQL": 90, "HTML": 55}
passed = {subject: mark for subject, mark in marks.items() if mark >= 60}
print(passed)
```

## 3. Set Comprehension


```python
numbers = [1, 2, 3, 4, 5]
squares = {x ** 2 for x in numbers}
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = {x for x in numbers if x % 2 == 0}
print(even_numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
result = {"Even" if x % 2 == 0 else "Odd" for x in numbers}
print(result)
```

```python
numbers = [1, 2, 2, 3, 3, 4, 4, 5]
unique_numbers = {x for x in numbers}
print(unique_numbers)
```

## 4. Nested Comprehensions


```python
matrix = [[1, 2], [3, 4], [5, 6]]
result = [[x * 2 for x in row] for row in matrix]
print(result)
```

```python
numbers = [1, 2, 3]
result = {
    x: {y: x * y for y in numbers}
    for x in numbers
}
print(result)
```

## 5. Points to Remember

- List comprehension creates a list.
- Dictionary comprehension creates key-value pairs.
- Set comprehension removes duplicate values.
- `if` can filter values.
- `if-else` can choose between values.
- Multiple `for` loops are possible.
- Use normal loops when comprehension logic becomes complex.
- Set elements do not have a guaranteed order.
