# Loops

Loops are used to **repeat a block of code multiple times**.

Instead of writing the same code again and again, we can use a loop.

````python
print(1)
print(2)
print(3)
print(4)
print(5)
````

Using a loop:

````python
for i in range(1, 6):
    print(i)
````

The loop makes the code shorter, cleaner, and easier to maintain.

# Types of Loops

Python has two main types of loops:

1. **`while` loop**
2. **`for` loop**

# 1. While Loop

A `while` loop executes a block of code **as long as a condition is `True`**.

## Syntax

`while` condition:

    statement

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

Here:

* `i = 1` initializes the variable.
* `i <= 5` is the condition.
* `print(i)` prints the current value.
* `i += 1` increases `i` by `1`.

````python
i = 5

while i >= 0:
    print(i)
    i -= 1
````

# Infinite Loop

If the condition never becomes `False`, the loop will continue forever.

````python
'''

i = 1

while i <= 5:

    print(i)

'''
````

This is an infinite loop because `i` never changes.

The condition will always remain: `1 <= 5`

So the loop never stops.

## Sum of Digits

````python
n = int(input("Enter a number: "))

total = 0

while n > 0:
    digit = n % 10
    total += digit
    n //= 10

print(total)
````

## Reverse a Number

````python
rev = int(input('Enter a number:'))
reverse = 0
while rev > 0:
    digit = rev%10
    reverse = reverse*10 + digit
    rev = rev//10
print(f"After reverse: {reverse}")
````

# 2. For Loop

A `for` loop is used when you want to repeat something for every item in a sequence or for a specific number of times.

## Syntax

`for` variable `in` sequence:

    statement

The variable takes one value at a time from the sequence.

````python
for i in range(5):
    print(i)
````

* `i` is the loop variable.
* `range(5)` generates numbers from `0` to `4`.
* The loop executes **5 times**.

````python
for i in range(10, 0, -1):
    print(i)
````

# Looping Through a String

A `for` loop can iterate through each character of a string.

````python
name = "Zee"

for char in name:
    print(char)
````

````python
name1 = 'Zeeshan Jamshed'
````

````python
for i in name1[::-1]:
    print(i)
````

# Looping Through a List

````python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
````

## Multiplication Table

````python
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
````

# Loop Control Statements

Python provides three important statements for controlling loops:

* `break`
* `continue`
* `pass`

These can be used inside both `for` and `while` loops.

## `break` in a `for` Loop

`break` is used to **stop the loop immediately**.

As soon as Python encounters `break`, the loop ends.

````python
for i in range(1, 100):
    if i == 5:
        break
    print(i)
````

When `i` becomes `5`, `break` stops the entire loop.

The numbers `5` to `99` are never printed.

## Find a Number

````python
num = [10, 20, 30, 40, 50, 60]
for i in num:
    if i == 30:
        print(i)
        break
````

````python
num = [10, 20, 30, 40, 50, 60]
for i in num:
    if i == 30:
        print('Number found')
        break
    print(i)
````

# `continue` in a `for` Loop

`continue` is used to **skip the current iteration**.

It does not stop the entire loop.

````python
for i in range(1, 6):
    if i == 3:
        continue

    print(i)
````

When `i` becomes `3`, Python skips the remaining code for that iteration and moves to the next iteration.

## Print Only Odd Numbers

````python
for i in range (1, 20):
    if i%2 == 0:
        continue
    print(i)
````

## `pass` in a `for` Loop

`pass` does **nothing**.

It is used as a placeholder when you want to write the structure of your code but don't want to add the actual logic yet.

````python
for i in range(5):
    pass
````

The loop runs five times, but nothing is displayed.

````python
for i in range(1, 6):
    if i == 3:
        pass

    print(i)
````

`pass` does nothing, so `3` is still printed.

````python
for i in range(1, 6):
    if i == 3:
        continue

    print(i)
````

`continue` skips the rest of the current iteration.

## For Loop With `else`

Python allows an `else` block with a `for` loop.

The `else` block executes when the loop finishes **normally**, without being stopped by `break`.

## Syntax

```python
for variable in sequence:
    statement
else:
    statement
```


````python
for i in range(5):
    print(i)
else:
    print("Loop completed")
````

## `for-else` With `break`

If the loop is stopped using `break`, the `else` block does not execute.

````python
for i in range(5):
    if i == 3:
        break

    print(i)
else:
    print("Loop completed")
````

The loop was terminated by `break`, so the `else` block was skipped.

## Searching With `for-else`

A common use of `for-else` is searching for an item.

````python
numbers = [10, 20, 30, 40, 50]

search = 30

for num in numbers:
    if num == search:
        print("Number found")
        break
else:
    print("Number not found")
````

````python
numbers = [10, 20, 30, 40, 50]

search = 60

for num in numbers:
    if num == search:
        print("Number found")
        break
else:
    print("Number not found")
````

# Nested Loops

````python
for i in range(1, 3):
    for j in range(10, 15):
        print(j, end = ' ')
    print('\n')
````

# Patterns Using Loops

A pattern usually consists of:

* Rows
* Columns
* Characters or numbers

For example:

```text
*
* *
* * *
* * * *
* * * * *
```

There are `5` rows.

The number of stars increases by `1` in every row.

````python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")

    print()
````

## Pattern 1: Increasing Star Pattern

````python
for i in range(1, 5):
    for j in range(i):
        print('*', end = " ")
    print()
````

## Pattern 2: Decreasing Star Pattern

````python
for i in range(5, 0, -1):
    for j in range(i):
        print('*', end=' ')
    print()
````

## Pattern 3: Number Pattern

````python
for i in range(1, 7):
    for j in range(1, i+1):
        print(j, end = " ")
    print()
````

## Pattern 4: Repeated Number Pattern

````python
for i in range(1, 7):
    for j in range(1, i+1):
        print(i, end=" ")
    print()
````

## Pattern 5: Continuous Number Pattern

````python
num = 0
for i in range(1, 7):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
````

## Pattern 6: Square Pattern

````python
for i in range(5):
    for j in range(5):
        print('*', end=' ')
    print()
````

## Pattern 7: Rectangle Pattern

````python
for i in range(5):
    for j in range(3):
        print('*', end=' ')
    print()
````

````python
for i in range(3):
    for j in range(5):
        print('*', end=' ')
    print()
````

## Pattern 8: Right-Aligned Star Pattern

````python
for i in range(1, 6):
    for j in range(5 -i):
        print(' ', end=' ')
    for j in range(i):
        print('*', end=' ')
    print()
````

## Pattern 9: Pyramid Pattern

````python
for i in range(1, 6):
    for j in range(5 -i):
        print(' ', end=' ')
    for j in range(2*i-1):
        print('*', end=' ')
    print()
````
