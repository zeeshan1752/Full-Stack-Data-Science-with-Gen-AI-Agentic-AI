# Python Functions

Functions are used to group a set of statements into a reusable block of code.

They help organize programs and allow the same code to be called multiple times.


## 1. Creating and Calling a Function

A function is defined using the `def` keyword.

A function runs when its name is called.


````python
def greet():
    print("Good Evening Team")

greet()
````

### Calling a Function Multiple Times

The same function can be called whenever it is needed.


````python
def greet():
    print("Good Evening Team")
    print("Welcome to Python Functions")

greet()
greet()
greet()
````

## 2. Function Syntax

Basic syntax:

```python
def function_name(parameters):
    # function body
    statements
    return value
```

- `def` - keyword used to define a function
- `function_name` - name of the function
- `parameters` - optional inputs
- function body - statements executed when the function is called
- `return` - optional value sent back to the caller


## 3. Function Without Arguments

A function does not always need parameters.


````python
def welcome():
    print("Welcome to Python")

welcome()
````

## 4. Function with Arguments

Arguments are values passed to a function when it is called.

The names written in the function definition are called **formal parameters**.

The values supplied during the function call are called **actual arguments**.


````python
def add(x, y):
    # x and y are formal parameters
    print(x + y)

add(5, 6)       # 5 and 6 are actual arguments
````

### Error: Too Many Arguments

If a function expects two arguments but three are supplied, Python raises a `TypeError`.


````python
def add(x, y):
    print(x + y)

add(5, 6, 7)
# TypeError: add() takes 2 positional arguments but 3 were given
````

### Error: Missing Arguments

If a required argument is not supplied, Python raises a `TypeError`.


````python
def add(x, y):
    print(x + y)

add(5)
# TypeError: add() missing 1 required positional argument: 'y'
````

## 5. Print vs Return

`print()` displays a value.

`return` sends a value back to the caller so it can be stored, reused, or processed.


````python
def add_using_print(x, y):
    print(x + y)

result = add_using_print(5, 6)

print("Returned value:", result)
print("Returned type:", type(result))
````

````python
def add_using_return(x, y):
    return x + y

result = add_using_return(5, 6)

print("Returned value:", result)
print("Returned type:", type(result))
````

### Using the Returned Value

A returned value can be used in another expression.


````python
def add(x, y):
    return x + y

result = add(10, 20)

print(result)
print(result * 2)
````

## 6. Returning Multiple Values

A function can return multiple values separated by commas.

Python returns them together as a tuple.


````python
def add_sub(x, y):
    return x + y, x - y

result = add_sub(5, 6)

print(result)
print(type(result))
````

````python
def add_sub_mul(x, y):
    return x + y, x - y, x * y

addition, subtraction, multiplication = add_sub_mul(5, 6)

print(addition)
print(subtraction)
print(multiplication)
````

## 7. Multiple Functions

More than one function can be defined in the same program.

Each function can perform a separate task.


````python
def greet():
    print("Hello")

def add(x, y):
    print(x + y)

def sub(x, y):
    print(x - y)

greet()
add(5, 4)
sub(10, 2)
````

# Types of Function Arguments

Python supports different ways of passing arguments:

1. Positional arguments
2. Keyword arguments
3. Default arguments
4. Variable-length positional arguments - `*args`
5. Variable-length keyword arguments - `**kwargs`


## 8. Positional Arguments

In positional arguments, values are assigned to parameters according to their position.


````python
def person(name, age, mobile):
    print("Name:", name)
    print("Age:", age)
    print("Mobile:", mobile)

person("Nit", 23, 2345)
````

### Error: Wrong Number of Positional Arguments

The number of supplied arguments must match the required parameters.


````python
def person(name, age, mobile):
    print(name, age, mobile)

person("Nit", 23)
# TypeError: missing 1 required positional argument: 'mobile'
````

````python
def person(name, age, mobile):
    print(name, age, mobile)

person("Nit", 23, 2345, "Extra")
# TypeError: takes 3 positional arguments but 4 were given
````

## 9. Keyword Arguments

In keyword arguments, values are passed using parameter names.

The order does not have to match the function definition.


````python
def person(name, age, mobile, addr):
    print("Name:", name)
    print("Age:", age)
    print("Mobile:", mobile)
    print("Address:", addr)

person(age=23, name="Nit", mobile=2345, addr="Hyd")
````

### Error: Unexpected Keyword

The keyword name must match a parameter name.


````python
def person(name, age, mobile, addr):
    print(name, age, mobile, addr)

person(age=23, name="Nit", mobile=2345, Addr="Hyd")
# TypeError: unexpected keyword argument 'Addr'
````

### Positional Argument After Keyword Argument

A positional argument cannot normally be placed after a keyword argument.


````python
def person(name, age, mobile):
    print(name, age, mobile)

person(name="Nit", 23, mobile=2345)
# SyntaxError: positional argument follows keyword argument
````

## 10. Default Arguments

A default argument gives a parameter a value in the function definition.

If that argument is not supplied, the default value is used.


````python
def person(name, age=18):
    print("Name:", name)
    print("Age:", age)

person("Vote")
person("Zeeshan", 22)
````

### Default Argument Rules

A non-default parameter cannot come after a default parameter.

Incorrect:

```python
def person(name="Guest", age):
    ...
```

This produces:

```text
SyntaxError: non-default argument follows default argument
```

Correct:

```python
def person(age, name="Guest"):
    ...
```


````python
def person(age, name="Guest"):
    print("Name:", name)
    print("Age:", age)

person(21)
person(21, "Zeeshan")
````

## 11. `*args` - Variable-Length Positional Arguments

`*args` is used when we do not know how many positional arguments will be passed.

Inside the function, `args` is a tuple.


````python
def add_all(*args):
    print("args:", args)
    print("Type:", type(args))

    total = 0
    for value in args:
        total += value

    return total

print(add_all(10, 20))
print(add_all(10, 20, 30, 40))
print(add_all())
````

### Keyword-Only Parameters with `*args`

A parameter written after `*args` becomes keyword-only.

So `x` must be supplied using its name.


````python
def example(*args, x):
    print("args:", args)
    print("x:", x)

# example(1, 2, 3)
# TypeError: missing 1 required keyword-only argument: 'x'

example(1, 2, 3, x=10)
````

### Unpacking with `*`

A list or tuple can be unpacked into separate positional arguments.


````python
numbers = [10, 20, 30, 40]

def add_four(a, b, c, d):
    return a + b + c + d

print(add_four(*numbers))
````

## 12. `**kwargs` - Variable-Length Keyword Arguments

`**kwargs` is used when we do not know how many keyword arguments will be passed.

Inside the function, `kwargs` is a dictionary.


````python
def student_details(**kwargs):
    print("kwargs:", kwargs)
    print("Type:", type(kwargs))

student_details(name="Zeeshan", age=22)
student_details(name="Zeeshan", age=22, city="Lucknow", course="B.Tech CSE")
````

### Unpacking with `**`

A dictionary can be unpacked into keyword arguments.


````python
details = {
    "name": "Zeeshan",
    "age": 22
}

def person(name, age):
    print("Name:", name)
    print("Age:", age)

person(**details)
````

## 13. `*args` vs `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Accepts | Extra positional arguments | Extra keyword arguments |
| Inside function | Tuple | Dictionary |
| Example | `func(10, 20)` | `func(a=10, b=20)` |


````python
def show(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

show(10, 20, 30, name="Zeeshan", age=22)
````

## 14. Combining Argument Types

A function can combine normal parameters, default parameters, `*args`, keyword-only parameters, and `**kwargs`.

A common order is:

```text
normal parameters
default parameters
*args
keyword-only parameters
**kwargs
```


````python
def demo(a, b=20, *args, city="Lucknow", **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("city:", city)
    print("kwargs:", kwargs)

demo(10, 30, 40, 50, city="Hyderabad", course="B.Tech CSE")
````

# Scope of Variables

Scope determines where a variable can be accessed.

Python follows the **LEGB** lookup rule:

1. Local
2. Enclosing
3. Global
4. Built-in


## 15. Local Scope

A variable created inside a function normally belongs to that function.

It cannot be directly accessed outside the function.


````python
def show():
    x = 10
    print("Inside function:", x)

show()

print(x)
# NameError: name 'x' is not defined
````

## 16. Global Scope

A variable created outside a function has global scope.

A function can read a global variable.


````python
x = 100

def show():
    print("Inside function:", x)

show()
print("Outside function:", x)
````

### Local and Global Variables with the Same Name

If a local variable has the same name as a global variable, the local variable is used inside the function.


````python
x = 100

def show():
    x = 50
    print("Local x:", x)

show()
print("Global x:", x)
````

## 17. `global` Keyword

The `global` keyword allows a function to modify a global variable.


````python
x = 10

def change():
    global x
    x = 50

change()
print(x)
````

### Error Without `global`

If Python sees an assignment to a name inside a function, it treats that name as local.

Trying to read that local variable before assigning it causes `UnboundLocalError`.


````python
x = 10

def change():
    # print(x)   # UnboundLocalError
    x = 20
    print("Local x:", x)

change()
print("Global x:", x)
````

## 18. `globals()` Function

`globals()` returns a dictionary representing the current global namespace.


````python
x = 100
name = "Zeeshan"

print(globals()["x"])
print(globals()["name"])
print(type(globals()))
````

## 19. Enclosing Scope and `nonlocal`

An enclosing variable belongs to an outer function.

The `nonlocal` keyword allows an inner function to modify a variable from its enclosing function.


````python
def outer():
    x = 10

    def inner():
        nonlocal x
        x += 5
        print("Inside inner:", x)

    inner()
    print("Inside outer:", x)

outer()
````

## 20. Built-in Scope

Python provides built-in names such as `print`, `len`, `sum`, `max`, and `type`.

They are available without defining them ourselves.


````python
numbers = [10, 20, 30]

print(len(numbers))
print(sum(numbers))
print(max(numbers))
print(type(numbers))
````

## 21. LEGB Rule Example

Python searches for a variable in this order:

**Local → Enclosing → Global → Built-in**


````python
x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        x = "Local"
        print(x)

    inner()

outer()
````

# Mutable and Immutable Objects in Functions

Python passes object references to functions, often described as **call by sharing**.

The behavior becomes easier to understand by comparing immutable and mutable objects.


## 22. Immutable Object Example

Integers are immutable.

Reassigning the parameter inside the function does not change the original integer.


````python
def change_number(x):
    x = x + 10
    print("Inside function:", x)

number = 50

change_number(number)

print("Outside function:", number)
````

## 23. Mutable Object Example

Lists are mutable.

A function can modify the contents of the same list object.


````python
def add_item(items):
    items.append(40)

numbers = [10, 20, 30]

add_item(numbers)

print(numbers)
````

### Reassigning a Mutable Parameter

Reassigning the parameter to a new list does not replace the caller's original list.


````python
def replace_list(items):
    items = [100, 200]
    print("Inside function:", items)

numbers = [10, 20, 30]

replace_list(numbers)

print("Outside function:", numbers)
````

# Nested and Higher-Order Functions

Functions are objects in Python. They can be stored in variables, passed as arguments, and returned from other functions.


## 24. Nested Functions

A function defined inside another function is called a nested function.

The inner function can access variables from the outer function.


````python
def outer():
    message = "Hello from outer function"

    def inner():
        print(message)

    inner()

outer()
````

## 25. Function as an Object

A function can be stored in another variable and called using that variable.


````python
def greet():
    print("Good Evening")

message = greet

message()
````

## 26. Passing a Function as an Argument

A function can be passed as an argument to another function.


````python
def square(x):
    return x * x

def calculate(func, value):
    return func(value)

print(calculate(square, 5))
````

## 27. Returning a Function

A function can return another function.


````python
def outer():
    def inner():
        return "Hello from inner function"

    return inner

result = outer()

print(result())
````

# Recursion

Recursion is a technique in which a function calls itself.

A recursive function normally contains:

- **Base case** - stops recursion
- **Recursive case** - calls the function again

Recursion is covered in detail in the separate `Recursion` topic.


````python
def countdown(n):
    if n == 0:
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

# Common Function Mistakes

### 1. Calling a function before defining it

```python
greet()

def greet():
    print("Hello")
```

This produces:

```text
NameError: name 'greet' is not defined
```

### 2. Forgetting to call the function

Writing `greet` refers to the function object. Writing `greet()` calls it.

### 3. Forgetting `return`

If a function has no `return` statement, its return value is `None`.


````python
def greet():
    print("Hello")

result = greet()

print("Returned value:", result)
print("Returned type:", type(result))
````

## 28. Function Documentation - Docstring

A docstring explains what a function does.

It is written as a string immediately after the function definition.


````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add(10, 20))
print(add.__doc__)
````

# Practical Function Examples

The following examples combine the concepts learned above.


## 29. Even or Odd Function


````python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    return "Odd"

number = 17
print(check_even_odd(number))
````

## 30. Maximum Value using `*args`


````python
def find_max(*numbers):
    if not numbers:
        return None

    maximum = numbers[0]

    for number in numbers[1:]:
        if number > maximum:
            maximum = number

    return maximum

print(find_max(10, 50, 20, 80, 30))
print(find_max())
````

## 31. Student Details using `**kwargs`


````python
def student(**details):
    for key, value in details.items():
        print(key, ":", value)

student(
    name="Zeeshan",
    course="B.Tech CSE",
    semester=7
)
````
