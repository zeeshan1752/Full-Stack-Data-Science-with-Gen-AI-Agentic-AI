# Exception Handling in Python

This notebook explains errors, exceptions, `try`, `except`, `except Exception as e`, `else`, `finally`, `raise`, custom exceptions, and practical examples.

## 1. Syntax, Logical and Runtime Errors

A syntax error prevents the code from being parsed. A logical error produces an incorrect result. A runtime error occurs while the program is executing.

````python
print('Hello'
````

The previous cell produces a `SyntaxError`. Fix the syntax before execution can continue.

````python
a = 10
b = 5
result = a - b
print('Result:', result)
````

The code runs, but if the intended operation was addition, this is a logical error. Python does not automatically know the intended logic.

````python
a = 10
b = 0
print(a / b)
````

The previous cell produces `ZeroDivisionError` because division by zero is not allowed.

## 2. Handling a Runtime Exception

````python
try:
    a = 10
    b = 0
    print(a / b)
except ZeroDivisionError:
    print('Cannot divide by zero')

print('Program continues')
````

Now the exception is handled and the program continues.

## 3. ValueError

````python
try:
    number = int('hello')
    print(number)
except ValueError as e:
    print('Error:', e)
````

### Why `as e`?

`e` stores the exception object, so we can inspect or print the actual error message.

````python
try:
    number = int('hello')
except Exception as e:
    print('Exception type:', type(e).__name__)
    print('Exception message:', e)
````

## 4. Specific Exception vs Generic Exception

````python
try:
    number = int('abc')
except ValueError as e:
    print('Specific handling:', e)
````

````python
try:
    number = int('abc')
except Exception as e:
    print('Generic handling:', e)
````

Specific exceptions are preferred when we know what problem we expect. `except Exception as e` is useful when we need a general handler or want to inspect an unexpected application exception.

## 5. Multiple `except` Blocks

````python
try:
    a = int(input('Enter first number: '))
    b = int(input('Enter second number: '))
    print('Result:', a / b)
except ValueError:
    print('Please enter integers only')
except ZeroDivisionError:
    print('Second number cannot be zero')
````

Try inputs such as `10` and `2`, `10` and `0`, or `abc` and `2` to see different handlers.

## 6. else Block

````python
try:
    number = int(input('Enter a number: '))
except ValueError:
    print('Invalid input')
else:
    print('Valid number:', number)
    print('The else block runs because no exception occurred.')
````

The `else` block executes only when the `try` block succeeds.

## 7. finally Block

````python
try:
    print('Opening resource')
    result = 10 / 0
except ZeroDivisionError:
    print('An error occurred')
finally:
    print('Closing resource')
````

`finally` runs whether an exception occurs or not. It is commonly used for cleanup.

## 8. try-except-else-finally

````python
try:
    number = int(input('Enter a number: '))
    result = 100 / number
except ValueError:
    print('Please enter a valid integer')
except ZeroDivisionError:
    print('Number cannot be zero')
else:
    print('Result:', result)
finally:
    print('Program execution completed')
````

## 9. Common Built-in Exceptions

````python
examples = [
    ('ZeroDivisionError', '10 / 0'),
    ('ValueError', "int('abc')"),
    ('TypeError', "'10' + 5"),
    ('NameError', 'print(age)'),
    ('IndexError', '[10, 20, 30][5]'),
    ('KeyError', "{'name': 'Zee'}['age']"),
    ('FileNotFoundError', "open('missing.txt')")
]

for name, example in examples:
    print(f'{name}: {example}')
````

## 10. File Handling and finally

````python
try:
    file = open('missing_file.txt', 'r')
    content = file.read()
    print(content)
except FileNotFoundError as e:
    print('File not found:', e)
finally:
    print('File operation completed')
````

## 11. raise

`raise` lets us create an exception intentionally when a business or validation rule is violated.

````python
try:
    age = -5
    if age < 0:
        raise ValueError('Age cannot be negative')
except ValueError as e:
    print('Validation error:', e)
````

## 12. Custom Exception

````python
class InvalidAgeError(Exception):
    pass

try:
    age = 15
    if age < 18:
        raise InvalidAgeError('Age must be 18 or above')
except InvalidAgeError as e:
    print('Custom exception:', e)
````

## 13. Database-style Example

In database code, cleanup is important. A connection should be closed even when a query fails.

````python
connection = None

try:
    print('Connecting to database')
    print('Executing query')
    # database query would be executed here
except Exception as e:
    print('Database error:', e)
finally:
    if connection is not None:
        connection.close()
    print('Database connection closed')
````

## 15. Best Practices

- Catch specific exceptions whenever possible.
- Use `as e` when the exception message is useful.
- Keep the `try` block focused.
- Use `else` for code that depends on successful execution.
- Use `finally` for cleanup.
- Do not silently ignore exceptions.
- Use `raise` for validation and business rules.
- Use custom exceptions when application-specific errors need clear names.

## 16. Points to Remember

`try` → risky code  
`except` → handles an exception  
`except Exception as e` → catches a normal exception and stores it in `e`  
`else` → runs when no exception occurs  
`finally` → runs whether an exception occurs or not  
`raise` → manually raises an exception

````python
try:
    risky_operation = 10 / 2
except Exception as e:
    print('Error:', e)
else:
    print('No error occurred. Result:', risky_operation)
finally:
    print('Cleanup completed')
````
