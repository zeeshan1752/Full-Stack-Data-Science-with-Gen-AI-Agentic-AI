## 1. What is Ollama?

- An **LLM** is the language model that generates a response.
- **Ollama** is the tool that helps you download and run supported models and exposes an interface for applications.
- The model, Ollama application, and Python client are three different things.

![Ollama workflow](images/ollama_workflow.png)

A model's answer can be incorrect or incomplete. Verify important information instead of assuming every response is true.

## 10. Send a prompt to a model

Change `model_name` to the exact model name shown by `ollama ls`. This cell sends one user message and prints the answer.


````python
# Replace this with a model name that is available on your computer.
model_name = "gemma3:270m"
question = "Explain artificial intelligence in simple words."

response = ollama.chat(
    model=model_name,
    messages=[{"role": "user", "content": question}],
)

print(response["message"]["content"])
````

### How the code works

- `model_name` identifies the model Ollama should use.
- `question` stores the prompt sent by the user.
- `messages` represents the conversation; each message has a role and content.
- `ollama.chat(...)` sends the request to the Ollama service.
- `response["message"]["content"]` extracts the generated text.

There is no fixed expected answer because the model generates the response.

## 12. Troubleshooting

| Error or issue | First thing to check |
|---|---|
| `ollama` command not found | Installation, terminal restart, and PATH. |
| No models listed | Download a model with `ollama pull MODEL_NAME`. |
| Model not found | Use the exact name/tag from `ollama ls`. |
| Python package missing | Run `python -m pip install -r requirements.txt` in the notebook's environment. |
| Python cannot connect | Make sure the Ollama service is running. |
| Model is slow or memory errors occur | Try a smaller model and close other resource-heavy applications. |

For more details, see the troubleshooting section in `README.md` and the [official Ollama documentation](https://docs.ollama.com/).
